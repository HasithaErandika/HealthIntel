```python
# ======================
# Step 1: Create age_cat from _AGE_G, _AGEG5YR, _AGE65YR
# ======================
def create_age_cat(row):
    if pd.notna(row.get('_AGE_G')) and row['_AGE_G'] in [1, 2, 3, 4, 5, 6]:
        return int(row['_AGE_G'])
    elif pd.notna(row.get('_AGEG5YR')):
        if row['_AGEG5YR'] == 1:
            return 1  # 18-24
        elif row['_AGEG5YR'] in [2, 3]:
            return 2  # 25-34
        elif row['_AGEG5YR'] in [4, 5]:
            return 3  # 35-44
        elif row['_AGEG5YR'] in [6, 7]:
            return 4  # 45-54
        elif row['_AGEG5YR'] in [8, 9]:
            return 5  # 55-64
        elif row['_AGEG5YR'] in [10, 11, 12, 13, 14]:
            return 6  # 65+
    elif pd.notna(row.get('_AGE65YR')):
        if row['_AGE65YR'] == 1:
            return 5  # 55-64
        elif row['_AGE65YR'] == 2:
            return 6  # 65+
    return np.nan

df['age_cat'] = df.apply(create_age_cat, axis=1)

# ======================
# Step 2: Create asthma_age_cat
# ======================
def create_asthma_age_cat(age):
    if pd.isna(age) or age in [97, 99]:
        return 'Missing'
    elif age < 30:
        return 'Early'
    elif 30 <= age <= 49:
        return 'Adult'
    elif 50 <= age <= 64:
        return 'Late_Adult'
    else:
        return 'Senior'

df['asthma_age_cat'] = df['ASTHMAGE'].apply(create_asthma_age_cat)

# ======================
# Step 3: Create diabetes_age_cat
# ======================
def create_diabetes_age_cat(age):
    if pd.isna(age) or age == 99:
        return 'Missing'
    elif age < 30:
        return 'Early'
    elif 30 <= age <= 49:
        return 'Adult'
    elif 50 <= age <= 64:
        return 'Late_Adult'
    else:
        return 'Senior'

df['diabetes_age_cat'] = df['DIABAGE2'].apply(create_diabetes_age_cat)

# ======================
# Step 4: Create household_size_cat
# ======================
def create_household_size_cat(size):
    if pd.isna(size) or size == 9:
        return 'Missing'
    elif size == 1:
        return 'Single'
    elif size == 2:
        return 'Standard'
    else:
        return 'Large'

df['household_size_cat'] = df['NUMHHOL2'].apply(create_household_size_cat)

# ======================
# Step 5: Display value counts
# ======================
categorical_cols = ['age_cat', 'asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']

for col in categorical_cols:
    vc = df[col].value_counts(dropna=False).sort_index()
    vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
    vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
    print(f"\nValue Counts for {col}:")
    print(vc_df)

# ======================
# Step 6: Visualization
# ======================
plt.style.use('seaborn-v0_8')  # or 'seaborn'

# Set up figure with 3 rows, 3 columns
fig = plt.figure(figsize=(18, 15))
gs = fig.add_gridspec(3, 3, height_ratios=[1, 1, 2])

# Bar plots (first row)
bar_cols = ['age_cat', 'asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']
for i, col in enumerate(bar_cols):
    row = 0
    col_idx = i if i < 3 else i - 3  # wrap to next column if more than 3
    ax = fig.add_subplot(gs[row, col_idx])
    sns.countplot(data=df, x=col, ax=ax, order=df[col].value_counts().index)
    ax.set_title(f'Bar Plot of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=45)

# Box plots (second row)
box_cols = ['asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']
for i, col in enumerate(box_cols):
    ax = fig.add_subplot(gs[1, i])
    sns.boxplot(data=df, x=col, y='age_cat', ax=ax)
    ax.set_title(f'Box Plot of age_cat by {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('age_cat')
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# ======================
# Step 7: Save updated dataset
# ======================
# Ensure `all_age_features` and `new_features` are defined
all_age_features = ['_AGEG5YR', '_AGE65YR', '_AGE_G', 'ASTHMAGE', 'DIABAGE2', 'NUMHHOL2']
new_features = ['age_cat', 'asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']

df = df.drop(columns=all_age_features, errors='ignore')
updated_path = '../data/processed/BRFSS_2.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
data_path = '../data/processed/BRFSS_3.csv'
df = pd.read_csv(data_path)

# Define general health features
general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3'],
    'Quality_of_Life': ['QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'PAFREQ1_', 'PAFREQ2_', 'QLSTRES2'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH']
}

# Flatten and keep only features present in df
all_general_health_features = (
    general_health_features['Standard']
    + general_health_features['Quality_of_Life']
    + general_health_features['Mental_Health']
)
all_general_health_features = df.columns.intersection(all_general_health_features).tolist()

if not all_general_health_features:
    raise ValueError("No general health features found in the dataset.")

# BRFSS-specific cleaning function
def clean_brfss_values(series):
    name = series.name
    if name in ['PHYSHLTH', 'MENTHLTH', 'POORHLTH']:
        # Days: 1-30 valid, 77=Don't know, 88=None, 99=Refused
        series = series.replace({77: np.nan, 99: np.nan, 88: 0})
    elif name in ['GENHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3',
                  'QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'QLSTRES2', 'HADSIGM3']:
        # Categorical: 7/9 = missing
        series = series.replace({7: np.nan, 9: np.nan})
    elif name in ['PAFREQ1_', 'PAFREQ2_']:
        # Physical activity: 99999 = refused
        series = series.replace({99999: np.nan})
    return series

# Apply cleaning
df[all_general_health_features] = df[all_general_health_features].apply(clean_brfss_values)

# Remove features that are ALL NaN after cleaning
valid_features = []
for col in all_general_health_features:
    if df[col].notna().any():
        valid_features.append(col)
    else:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All general health features are NaN after cleaning.")

# Update feature lists
numeric_features = [f for f in valid_features if pd.api.types.is_numeric_dtype(df[f])]
categorical_features = [f for f in valid_features if f not in numeric_features]

# -------------------------------
# Summary Statistics (unchanged)
# -------------------------------
print("Summary statistics for all general health features:\n")
for feature in valid_features:
    print(f"\nFeature: {feature}")
    if pd.api.types.is_numeric_dtype(df[feature]):
        print(df[feature].describe())
    else:
        counts = df[feature].value_counts(dropna=False)
        percentages = (counts / len(df) * 100).round(2)
        print(pd.DataFrame({'Count': counts, 'Percentage': percentages}))

# -------------------------------
# Visualization with 3x3 grids
# -------------------------------
plt.style.use('seaborn-v0_8')

# --- Numeric Features: Boxplots in 3x3 grids ---
if numeric_features:
    n_numeric = len(numeric_features)
    n_cols = 3
    n_rows = (n_numeric + n_cols - 1) // n_cols  # Ceiling division

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_numeric > 1 else [axes]

    for i, feature in enumerate(numeric_features):
        sns.boxplot(y=df[feature], ax=axes[i])
        axes[i].set_title(f'Boxplot of {feature}')
        axes[i].set_ylabel(feature)

    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()

    # --- Histograms in separate 3x3 grids ---
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_numeric > 1 else [axes]

    for i, feature in enumerate(numeric_features):
        sns.histplot(df[feature].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {feature}')
        axes[i].set_xlabel(feature)

    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()

# --- Categorical Features: Barplots in 3x3 grids ---
if categorical_features:
    n_cat = len(categorical_features)
    n_cols = 3
    n_rows = (n_cat + n_cols - 1) // n_cols

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_cat > 1 else [axes]

    for i, feature in enumerate(categorical_features):
        order = df[feature].value_counts(dropna=False).index
        sns.countplot(data=df, x=feature, order=order, ax=axes[i])
        axes[i].set_title(f'Bar Plot of {feature}')
        axes[i].set_xlabel(feature)
        axes[i].tick_params(axis='x', rotation=45)

    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



# Load the data
data_path = '../data/processed/BRFSS_3.csv'

df = pd.read_csv(data_path)

# Define general health features
general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3'],
    'Quality_of_Life': ['QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'PAFREQ1_', 'PAFREQ2_', 'QLSTRES2'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH']
}

# Flatten and filter valid features
all_general_health_features = (
    general_health_features['Standard']
    + general_health_features['Quality_of_Life']
    + general_health_features['Mental_Health']
)
all_general_health_features = df.columns.intersection(all_general_health_features).tolist()

if not all_general_health_features:
    raise ValueError("No general health features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_values(series):
    if series.name in ['PHYSHLTH', 'MENTHLTH', 'POORHLTH']:
        return series.where(~series.isin([77, 99]), np.nan).replace(88, 0)
    elif series.name in ['GENHLTH']:
        return series.where(~series.isin([7, 9]), np.nan)
    elif series.name in ['QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3', 'HADSIGM3', '_RFHLTH']:
        return series.where(~series.isin([7, 9]), np.nan)
    elif series.name in ['PAFREQ1_', 'PAFREQ2_']:
        return series.where(~series.isin([99999]), np.nan)
    return series

df[all_general_health_features] = df[all_general_health_features].apply(clean_brfss_values)

# ======================
# Step 2: Engineer Standard features
# ======================
# Binary: Good or better (1-3) vs. Fair or poor (4-5)
df['general_health_cat'] = np.where(df['GENHLTH'].isin([1, 2, 3]), 1, 
                                   np.where(df['GENHLTH'].isin([4, 5]), 0, np.nan))

# Binary: 1+ days vs. 0 days
df['physical_health_bad'] = np.where(df['PHYSHLTH'] >= 1, 1, 
                                    np.where(df['PHYSHLTH'] == 0, 0, np.nan))

df['poor_health_days'] = np.where(df['POORHLTH'] >= 1, 1, 
                                  np.where(df['POORHLTH'] == 0, 0, np.nan))

# Binary: Yes (1) vs. No (2)
def binarize_brfss(col):
    return np.where(col == 1, 1, np.where(col == 2, 0, np.nan))

df['has_activity_limitation'] = binarize_brfss(df['QLACTLM2'])
df['has_depression'] = binarize_brfss(df['ADDEPEV2'])
df['has_asthma'] = binarize_brfss(df['ASTHMA3'])
df['has_arthritis'] = binarize_brfss(df['HAVARTH3'])


# ======================
# Step 3: Engineer Quality_of_Life features
# ======================
# Exclude QLMENLT2, QLHLTH2, PAINACT2, QLSTRES2 (missing)
df['general_health_risk'] = binarize_brfss(df['_RFHLTH'])

# Standardize physical activity frequency to times per week
# BRFSS: 1XXX=weekly, 2XXX=monthly (divide by 4.33 to convert to weekly)
def standardize_activity_freq(value):
    if pd.isna(value):
        return np.nan
    if 1000 <= value <= 1999:
        return value - 1000
    elif 2000 <= value <= 2999:
        return (value - 2000) / 4.33
    return np.nan

df['vigorous_activity_freq'] = df['PAFREQ1_'].apply(standardize_activity_freq)
df['moderate_activity_freq'] = df['PAFREQ2_'].apply(standardize_activity_freq)

# ======================
# Step 4: Engineer Mental_Health features
# ======================
df['has_anxiety'] = binarize_brfss(df['HADSIGM3'])
df['mental_health_bad'] = np.where(df['MENTHLTH'] >= 1, 1, 
                                  np.where(df['MENTHLTH'] == 0, 0, np.nan))

# Composite: Anxiety or 1+ mental health bad days
def mental_health_condition(row):
    if pd.isna(row['has_anxiety']) or pd.isna(row['mental_health_bad']):
        return np.nan
    return 1 if row['has_anxiety'] == 1 or row['mental_health_bad'] == 1 else 0

df['mental_health_condition'] = df.apply(mental_health_condition, axis=1)

# Updated general_health_features dictionary
general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3',
                 'general_health_cat', 'physical_health_bad', 'poor_health_days', 'has_activity_limitation',
                 'has_depression', 'has_asthma', 'has_arthritis'],
    'Quality_of_Life': ['_RFHLTH', 'PAFREQ1_', 'PAFREQ2_', 'general_health_risk', 
                        'vigorous_activity_freq', 'moderate_activity_freq'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH', 'has_anxiety', 'mental_health_bad', 'mental_health_condition']
}

# ======================
# Step 5: Display value counts for categorical/binary features
# ======================
categorical_cols = ['general_health_cat', 'physical_health_bad', 'poor_health_days', 'has_activity_limitation',
                    'has_depression', 'has_asthma', 'has_arthritis',
                    'general_health_risk', 'has_anxiety', 'mental_health_condition']

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 6: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Bar plots for categorical/binary features
cols = 3
rows = (len(categorical_cols) + cols - 1) // cols
fig = plt.figure(figsize=(6 * cols, 5 * rows))

for i, col in enumerate(categorical_cols):
    if col in df.columns:
        ax = fig.add_subplot(rows, cols, i + 1)
        order = df[col].value_counts(dropna=False).index
        sns.countplot(data=df, x=col, order=order, ax=ax)
        ax.set_title(f'Bar Plot of {col}')
        ax.set_xlabel(col)
        ax.set_ylabel('Count')
        ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# ======================
# Step 7: Save updated dataset
# ======================
all_original_features = ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3',
                         'QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'PAFREQ1_', 'PAFREQ2_', 'QLSTRES2',
                         'HADSIGM3', 'MENTHLTH']
new_features = ['general_health_cat', 'physical_health_bad', 'poor_health_days', 'has_activity_limitation',
                'has_depression', 'has_asthma', 'has_arthritis',
                'general_health_risk', 'vigorous_activity_freq', 'moderate_activity_freq',
                'has_anxiety', 'mental_health_bad', 'mental_health_condition']

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_4.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



# Load the data
data_path = '../data/processed/BRFSS_4.csv'


df = pd.read_csv(data_path)

# Define chronic conditions features
chronic_conditions = {
    'Standard': ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'CHKHEMO3'],
    'Diabetes': ['DIABETE3', 'PREDIAB1', 'DOCTDIAB', 'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST'],
    'Hypertension': ['_RFHYPE5', 'BLDSUGAR']
}

# Flatten and keep only features present in df
all_chronic_features = (
    chronic_conditions['Standard']
    + chronic_conditions['Diabetes']
    + chronic_conditions['Hypertension']
)
all_chronic_features = df.columns.intersection(all_chronic_features).tolist()

if not all_chronic_features:
    raise ValueError("No chronic condition features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_values(series):
    name = series.name
    if name in ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'DIABETE3', 'PREDIAB1', 
                'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST', '_RFHYPE5']:
        return series.replace({7: np.nan, 9: np.nan})
    elif name in ['CHKHEMO3', 'DOCTDIAB']:
        return series.replace({98: np.nan, 99: np.nan, 88: 0})
    elif name == 'BLDSUGAR':
        return series.replace({999: np.nan})
    return series

df[all_chronic_features] = df[all_chronic_features].apply(clean_brfss_values)

# Remove features that are all NaN after cleaning
valid_features = [col for col in all_chronic_features if df[col].notna().any()]
for col in all_chronic_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All chronic condition features are NaN after cleaning.")

# ======================
# Step 2: Engineer Standard features
# ======================
# Binary features
df['has_high_bp'] = np.where(df['BPHIGH4'].isin([1, 4]), 1, 
                            np.where(df['BPHIGH4'] == 3, 0, np.nan))
df['on_bp_meds'] = np.where(df['BPMEDS'] == 1, 1, 
                            np.where(df['BPMEDS'] == 2, 0, np.nan))
df['has_high_cholesterol'] = np.where(df['TOLDHI2'] == 1, 1, 
                                     np.where(df['TOLDHI2'] == 2, 0, np.nan))
df['has_kidney_disease'] = np.where(df['CHCKIDNY'] == 1, 1, 
                                    np.where(df['CHCKIDNY'] == 2, 0, np.nan))
df['a1c_checks'] = np.where(df['CHKHEMO3'] >= 1, 1, 
                            np.where(df['CHKHEMO3'] == 0, 0, np.nan))

# Composite: Count of standard chronic conditions
def chronic_conditions_standard(row):
    conditions = [row['has_high_bp'], row['has_high_cholesterol'], row['has_kidney_disease']]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['chronic_conditions_standard'] = df.apply(chronic_conditions_standard, axis=1)

# ======================
# Step 3: Engineer Diabetes features
# ======================
df['has_diabetes'] = np.where(df['DIABETE3'] == 1, 1, 
                             np.where(df['DIABETE3'].isin([3, 4]), 0, np.nan))
df['has_prediabetes'] = np.where(df['PREDIAB1'] == 1, 1, 
                                np.where(df['PREDIAB1'] == 3, 0, np.nan))
df['diabetes_doctor_visits'] = np.where(df['DOCTDIAB'] >= 1, 1, 
                                       np.where(df['DOCTDIAB'] == 0, 0, np.nan))
df['diabetes_eye_exam'] = np.where(df['DIABEYE'] == 1, 1, 
                                  np.where(df['DIABEYE'] == 2, 0, np.nan))
df['uses_insulin'] = np.where(df['INSULIN'] == 1, 1, 
                             np.where(df['INSULIN'] == 2, 0, np.nan))
df['diabetes_education'] = np.where(df['DIABEDU'] == 1, 1, 
                                   np.where(df['DIABEDU'] == 2, 0, np.nan))
df['tested_for_diabetes'] = np.where(df['PDIABTST'] == 1, 1, 
                                    np.where(df['PDIABTST'] == 2, 0, np.nan))

# Composite: Diabetes management score
def diabetes_management(row):
    conditions = [row['diabetes_doctor_visits'], row['diabetes_eye_exam'], 
                  row['uses_insulin'], row['diabetes_education']]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['diabetes_management'] = df.apply(diabetes_management, axis=1)

# ======================
# Step 4: Engineer Hypertension features
# ======================
df['high_bp_risk'] = np.where(df['_RFHYPE5'] == 2, 1, 
                             np.where(df['_RFHYPE5'] == 1, 0, np.nan))

# Standardize BLDSUGAR to times per year
def standardize_blood_sugar(value):
    if pd.isna(value):
        return np.nan
    if 101 <= value <= 199:
        return (value - 100) * 52  # Weekly to yearly
    elif 201 <= value <= 299:
        return (value - 200) * 12  # Monthly to yearly
    elif 301 <= value <= 399:
        return value - 300  # Yearly
    elif value == 401:
        return 0  # Never
    return np.nan

df['blood_sugar_checks'] = df['BLDSUGAR'].apply(standardize_blood_sugar)
df['has_blood_sugar_checks'] = np.where(df['blood_sugar_checks'] >= 1, 1, 
                                       np.where(df['blood_sugar_checks'] == 0, 0, np.nan))

# Composite: Hypertension management
def hypertension_management(row):
    conditions = [row['on_bp_meds'], row['has_blood_sugar_checks']]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['hypertension_management'] = df.apply(hypertension_management, axis=1)

# Updated chronic_conditions dictionary
chronic_conditions = {
    'Standard': ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'CHKHEMO3', 
                 'has_high_bp', 'on_bp_meds', 'has_high_cholesterol', 'has_kidney_disease', 
                 'a1c_checks', 'chronic_conditions_standard'],
    'Diabetes': ['DIABETE3', 'PREDIAB1', 'DOCTDIAB', 'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST',
                 'has_diabetes', 'has_prediabetes', 'diabetes_doctor_visits', 'diabetes_eye_exam',
                 'uses_insulin', 'diabetes_education', 'tested_for_diabetes', 'diabetes_management'],
    'Hypertension': ['_RFHYPE5', 'BLDSUGAR', 'high_bp_risk', 'blood_sugar_checks', 
                     'has_blood_sugar_checks', 'hypertension_management']
}

# ======================
# Step 5: Display value counts for categorical/binary features
# ======================
categorical_cols = ['has_high_bp', 'on_bp_meds', 'has_high_cholesterol', 'has_kidney_disease',
                   'a1c_checks', 'chronic_conditions_standard', 'has_diabetes', 'has_prediabetes',
                   'diabetes_doctor_visits', 'diabetes_eye_exam', 'uses_insulin', 
                   'diabetes_education', 'tested_for_diabetes', 'diabetes_management',
                   'high_bp_risk', 'has_blood_sugar_checks', 'hypertension_management']

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 6: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Bar plots for categorical/binary features
n_cols = 3
n_cat = len(categorical_cols)
n_rows = (n_cat + n_cols - 1) // n_cols
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
axes = axes.flatten() if n_cat > 1 else [axes]

for i, col in enumerate(categorical_cols):
    if col in df.columns:
        order = df[col].value_counts(dropna=False).index
        sns.countplot(data=df, x=col, order=order, ax=axes[i])
        axes[i].set_title(f'Bar Plot of {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Count')
        axes[i].tick_params(axis='x', rotation=45)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()


# ======================
# Step 7: Save updated dataset
# ======================
all_original_features = ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'CHKHEMO3', 
                         'DIABETE3', 'PREDIAB1', 'DOCTDIAB', 'DIABEYE', 'INSULIN', 
                         'DIABEDU', 'PDIABTST', '_RFHYPE5', 'BLDSUGAR']
new_features = ['has_high_bp', 'on_bp_meds', 'has_high_cholesterol', 'has_kidney_disease', 
                'a1c_checks', 'chronic_conditions_standard', 'has_diabetes', 'has_prediabetes', 
                'diabetes_doctor_visits', 'diabetes_eye_exam', 'uses_insulin', 
                'diabetes_education', 'tested_for_diabetes', 'diabetes_management', 
                'high_bp_risk', 'has_blood_sugar_checks', 'hypertension_management']

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_5.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load the data
data_path = '../data/processed/BRFSS_7.csv'
df = pd.read_csv(data_path)

# Define care access features (from BRFSS codebook)
care_access_features = {
    'Standard': ['HLTHPLN1',  # Have any kind of health care coverage
                 'PERSDOC2',  # Have one person as personal doctor
                 'MEDCOST',   # Could not see doctor because of cost
                 'CHECKUP1',  # Last routine checkup
                 '_HCVU651'], # Health care coverage, age 18–64
    'Support_Services': ['EMTSUPRT',  # Emotional support
                         'USENOW3',   # Current smokeless tobacco use (corrected from prompt label)
                         'LMTJOIN3',  # Limited by joint symptoms
                         'VETERAN3']  # Ever served in U.S. Armed Forces
}

# Flatten and keep only features present in df
all_care_access_features = (
    care_access_features['Standard']
    + care_access_features['Support_Services']
)
all_care_access_features = df.columns.intersection(all_care_access_features).tolist()

if not all_care_access_features:
    raise ValueError("No care access features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_values(series):
    name = series.name

    # Yes/No/Other categorical (7 = Don't know, 9 = Refused → NaN)
    if name in ['HLTHPLN1', 'PERSDOC2', 'MEDCOST', '_HCVU651',
                'EMTSUPRT', 'USENOW3', 'LMTJOIN3', 'VETERAN3']:
        return series.replace({7: np.nan, 9: np.nan})

    # Last routine checkup (CHECKUP1): 1=Within past year, 2=Within past 2 years,
    # 3=Within past 5 years, 4=5+ years ago, 7=Don't know, 9=Refused
    elif name == 'CHECKUP1':
        return series.replace({7: np.nan, 9: np.nan})

    return series

# Apply cleaning
df[all_care_access_features] = df[all_care_access_features].apply(clean_brfss_values)

# Remove all-NaN features
valid_features = [col for col in all_care_access_features if df[col].notna().any()]
if not valid_features:
    raise ValueError("All care access features are NaN after cleaning.")

# ======================
# Step 2: Engineer Standard features
# ======================
# Binary: Has health coverage (1=Yes, 2=No)
df['has_health_coverage'] = np.where(df['HLTHPLN1'] == 1, 1, 
                                    np.where(df['HLTHPLN1'] == 2, 0, np.nan))

# Binary: Has personal doctor (1 or 2=Yes, 3=No)
df['has_personal_doctor'] = np.where(df['PERSDOC2'].isin([1, 2]), 1, 
                                    np.where(df['PERSDOC2'] == 3, 0, np.nan))

# Binary: Cost barrier to care (1=Yes, 2=No)
df['has_cost_barrier'] = np.where(df['MEDCOST'] == 1, 1, 
                                 np.where(df['MEDCOST'] == 2, 0, np.nan))

# Binary: Recent routine checkup (1 or 2=Within 2 years, 3/4/8=Longer or never)
df['recent_checkup'] = np.where(df['CHECKUP1'].isin([1, 2]), 1, 
                               np.where(df['CHECKUP1'].isin([3, 4, 8]), 0, np.nan))

# Binary: Has coverage for age 18-64 (1=Yes, 2=No)
df['has_coverage_18_64'] = np.where(df['_HCVU651'] == 1, 1, 
                                   np.where(df['_HCVU651'] == 2, 0, np.nan))

# Composite: Care access score (sum of positives: coverage, personal doctor, recent checkup, no cost barrier)
def care_access_score(row):
    conditions = [row['has_health_coverage'], row['has_personal_doctor'], 
                  row['recent_checkup'], (1 - row['has_cost_barrier'] if not pd.isna(row['has_cost_barrier']) else np.nan)]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['care_access_score'] = df.apply(care_access_score, axis=1)

# ======================
# Step 3: Engineer Support_Services features
# ======================
# Binary: Has adequate emotional support (1,2,3=Yes [Always/Usually/Sometimes], 4,5=No [Rarely/Never])
df['has_emotional_support'] = np.where(df['EMTSUPRT'].isin([1, 2, 3]), 1, 
                                      np.where(df['EMTSUPRT'].isin([4, 5]), 0, np.nan))

# Binary: Uses smokeless tobacco (1 or 2=Yes [Every day/Some days], 3=No)
df['uses_smokeless_tobacco'] = np.where(df['USENOW3'].isin([1, 2]), 1, 
                                       np.where(df['USENOW3'] == 3, 0, np.nan))

# Binary: Limited by joint symptoms (1=Yes, 2=No)
df['limited_by_joints'] = np.where(df['LMTJOIN3'] == 1, 1, 
                                  np.where(df['LMTJOIN3'] == 2, 0, np.nan))

# Binary: Is veteran (1=Yes, 2=No)
df['is_veteran'] = np.where(df['VETERAN3'] == 1, 1, 
                           np.where(df['VETERAN3'] == 2, 0, np.nan))

# Composite: Support services indicator (sum of has_emotional_support, is_veteran, (1 - limited_by_joints), (1 - uses_smokeless_tobacco))
def support_services_indicator(row):
    conditions = [row['has_emotional_support'], row['is_veteran'], 
                  (1 - row['limited_by_joints'] if not pd.isna(row['limited_by_joints']) else np.nan), 
                  (1 - row['uses_smokeless_tobacco'] if not pd.isna(row['uses_smokeless_tobacco']) else np.nan)]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['support_services_indicator'] = df.apply(support_services_indicator, axis=1)

# Updated care_access_features dictionary
care_access_features = {
    'Standard': ['HLTHPLN1', 'PERSDOC2', 'MEDCOST', 'CHECKUP1', '_HCVU651',
                 'has_health_coverage', 'has_personal_doctor', 'has_cost_barrier', 
                 'recent_checkup', 'has_coverage_18_64', 'care_access_score'],
    'Support_Services': ['EMTSUPRT', 'USENOW3', 'LMTJOIN3', 'VETERAN3',
                         'has_emotional_support', 'uses_smokeless_tobacco', 
                         'limited_by_joints', 'is_veteran', 'support_services_indicator']
}

# ======================
# Step 4: Display value counts for engineered features
# ======================
engineered_cols = ['has_health_coverage', 'has_personal_doctor', 'has_cost_barrier', 
                   'recent_checkup', 'has_coverage_18_64', 'care_access_score', 
                   'has_emotional_support', 'uses_smokeless_tobacco', 
                   'limited_by_joints', 'is_veteran', 'support_services_indicator']

for col in engineered_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 5: Visualization with 3x3 grids
# ======================
# Categorical/Engineered Binary Features Barplots
if engineered_cols:
    n_cols = 3
    n_rows = (len(engineered_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, feature in enumerate(engineered_cols):
        if feature in df.columns:
            order = df[feature].value_counts(dropna=False).index
            sns.countplot(x=df[feature], order=order, ax=axes[i])
            axes[i].set_title(f'Bar Plot of {feature}')
            axes[i].tick_params(axis='x', rotation=45)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()


# ======================
# Step 6: Save updated dataset
# ======================
all_original_features = ['HLTHPLN1', 'PERSDOC2', 'MEDCOST', 'CHECKUP1', '_HCVU651',
                         'EMTSUPRT', 'USENOW3', 'LMTJOIN3', 'VETERAN3']
new_features = ['has_health_coverage', 'has_personal_doctor', 'has_cost_barrier', 
                'recent_checkup', 'has_coverage_18_64', 'care_access_score', 
                'has_emotional_support', 'uses_smokeless_tobacco', 
                'limited_by_joints', 'is_veteran', 'support_services_indicator']

df = df.drop(columns=all_original_features, errors='ignore')
updated_path = '../data/processed/BRFSS_8.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load dataset
data_path = '../data/processed/BRFSS_8.csv'

df = pd.read_csv(data_path)

# Define Physical Activity features
physical_activity_features = {
    'Standard': ['_TOTINDA', 'EXERANY2', 'EXERHMM1', 'EXERHMM2',
                 '_PASTRNG', 'EXEROFT1', 'EXEROFT2', 'STRENGTH', 'STRFREQ_'],
    'Exercise_Duration': ['PADUR1_', 'PADUR2_'],
    'Fitness_Tools': ['MAXVO2_', 'STREHAB1', 'HAREHAB1', 'USEEQUIP']
}

# Flatten and keep only features present in df
all_pa_features = (
    physical_activity_features['Standard'] +
    physical_activity_features['Exercise_Duration'] +
    physical_activity_features['Fitness_Tools']
)
all_pa_features = df.columns.intersection(all_pa_features).tolist()

if not all_pa_features:
    raise ValueError("No physical activity features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_pa(series):
    name = series.name
    if name in ['_TOTINDA', 'EXERANY2', '_PASTRNG', 'STREHAB1', 'HAREHAB1', 'USEEQUIP']:
        return series.replace({7: np.nan, 9: np.nan})
    elif name == 'STRENGTH':
        return series.replace({999: np.nan, 888: 0})
    elif name in ['EXEROFT1', 'EXEROFT2']:
        return series.replace({777: np.nan, 999: np.nan})
    elif name == 'STRFREQ_':
        return series.replace({99900: np.nan})
    elif name in ['EXERHMM1', 'EXERHMM2', 'PADUR1_', 'PADUR2_']:
        return series.replace({7777: np.nan, 9999: np.nan})
    elif name == 'MAXVO2_':
        return series.replace({9999: np.nan})
    return series

df[all_pa_features] = df[all_pa_features].apply(clean_brfss_pa)

# Remove features that are entirely NaN
valid_features = [col for col in all_pa_features if df[col].notna().any()]
for col in all_pa_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All physical activity features are NaN after cleaning.")

# ======================
# Step 2: Engineer Standard features
# ======================
df['has_physical_activity'] = np.where(df['_TOTINDA'] == 1, 1, 
                                      np.where(df['_TOTINDA'] == 2, 0, np.nan))
df['exercised_past_30d'] = np.where(df['EXERANY2'] == 1, 1, 
                                    np.where(df['EXERANY2'] == 2, 0, np.nan))
df['primary_exercise_duration'] = df['EXERHMM1']
df['secondary_exercise_duration'] = df['EXERHMM2']
df['meets_strength_training'] = np.where(df['_PASTRNG'] == 1, 1, 
                                        np.where(df['_PASTRNG'] == 2, 0, np.nan))
df['has_strength_training'] = np.where(df['STRENGTH'] >= 1, 1, 
                                      np.where(df['STRENGTH'] == 0, 0, np.nan))

# Standardize exercise frequencies to times per week
def standardize_freq(value):
    if pd.isna(value):
        return np.nan
    if 101 <= value <= 199:
        return value - 100  # Weekly
    elif 201 <= value <= 299:
        return (value - 200) / 4.33  # Monthly to weekly
    return np.nan

df['primary_exercise_freq'] = df['EXEROFT1'].apply(standardize_freq)
df['secondary_exercise_freq'] = df['EXEROFT2'].apply(standardize_freq)
df['strength_training_freq'] = df['STRFREQ_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)

# Composite: Total exercise frequency per week
def total_exercise_freq(row):
    freqs = [row['primary_exercise_freq'], row['secondary_exercise_freq'], row['strength_training_freq']]
    if any(pd.isna(f) for f in freqs):
        return np.nan
    return sum(f for f in freqs if not pd.isna(f))

df['total_exercise_freq'] = df.apply(total_exercise_freq, axis=1)

# ======================
# Step 3: Engineer Exercise_Duration features
# ======================
df['primary_activity_duration'] = df['PADUR1_']
df['secondary_activity_duration'] = df['PADUR2_']

# Composite: Total activity duration
def total_activity_duration(row):
    durations = [row['primary_activity_duration'], row['secondary_activity_duration']]
    if any(pd.isna(d) for d in durations):
        return np.nan
    return sum(d for d in durations if not pd.isna(d))

df['total_activity_duration'] = df.apply(total_activity_duration, axis=1)

# ======================
# Step 4: Engineer Fitness_Tools features
# ======================
df['vo2_max'] = df['MAXVO2_']
df['in_strength_rehab'] = np.where(df['STREHAB1'] == 1, 1, 
                                  np.where(df['STREHAB1'] == 2, 0, np.nan))
df['in_heart_rehab'] = np.where(df['HAREHAB1'] == 1, 1, 
                                np.where(df['HAREHAB1'] == 2, 0, np.nan))
df['uses_exercise_equipment'] = np.where(df['USEEQUIP'] == 1, 1, 
                                        np.where(df['USEEQUIP'] == 2, 0, np.nan))

# Composite: Fitness program participation
def fitness_program_participation(row):
    conditions = [row['in_strength_rehab'], row['in_heart_rehab'], row['uses_exercise_equipment']]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['fitness_program_participation'] = df.apply(fitness_program_participation, axis=1)

# Updated physical_activity_features dictionary
physical_activity_features = {
    'Standard': ['_TOTINDA', 'EXERANY2', 'EXERHMM1', 'EXERHMM2', '_PASTRNG', 'EXEROFT1', 
                 'EXEROFT2', 'STRENGTH', 'STRFREQ_', 'has_physical_activity', 'exercised_past_30d', 
                 'primary_exercise_duration', 'secondary_exercise_duration', 'meets_strength_training', 
                 'primary_exercise_freq', 'secondary_exercise_freq', 'has_strength_training', 
                 'strength_training_freq', 'total_exercise_freq'],
    'Exercise_Duration': ['PADUR1_', 'PADUR2_', 'primary_activity_duration', 
                         'secondary_activity_duration', 'total_activity_duration'],
    'Fitness_Tools': ['MAXVO2_', 'STREHAB1', 'HAREHAB1', 'USEEQUIP', 'vo2_max', 
                      'in_strength_rehab', 'in_heart_rehab', 'uses_exercise_equipment', 
                      'fitness_program_participation']
}

# ======================
# Step 5: Display value counts for categorical/binary features
# ======================
categorical_cols = ['has_physical_activity', 'exercised_past_30d', 'meets_strength_training', 
                   'has_strength_training', 'in_strength_rehab', 'in_heart_rehab', 
                   'uses_exercise_equipment', 'fitness_program_participation']

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 6: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Numeric Features: Boxplots and Histograms
numeric_cols = ['primary_exercise_duration', 'secondary_exercise_duration', 'primary_exercise_freq', 
                'secondary_exercise_freq', 'strength_training_freq', 'total_exercise_freq', 
                'primary_activity_duration', 'secondary_activity_duration', 'total_activity_duration', 
                'vo2_max', 'fitness_program_participation']
valid_numeric_cols = [col for col in numeric_cols if col in df.columns]
if valid_numeric_cols:
    n_cols = 3
    n_rows = (len(valid_numeric_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(valid_numeric_cols):
        sns.boxplot(y=df[col], ax=axes[i])
        axes[i].set_title(f'Boxplot of {col}')
        axes[i].set_ylabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, col in enumerate(valid_numeric_cols):
        sns.histplot(df[col].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {col}')
        axes[i].set_xlabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

# Categorical Features: Barplots
if categorical_cols:
    n_cols = 3
    n_rows = (len(categorical_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(categorical_cols):
        if col in df.columns:
            order = df[col].value_counts(dropna=False).index
            sns.countplot(x=df[col], order=order, ax=axes[i])
            axes[i].set_title(f'Bar Plot of {col}')
            axes[i].set_xlabel(col)
            axes[i].tick_params(axis='x', rotation=45)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()


# ======================
# Step 7: Save updated dataset
# ======================
all_original_features = ['_TOTINDA', 'EXERANY2', 'EXERHMM1', 'EXERHMM2', '_PASTRNG', 
                         'EXEROFT1', 'EXEROFT2', 'STRENGTH', 'STRFREQ_', 'PADUR1_', 
                         'PADUR2_', 'MAXVO2_', 'STREHAB1', 'HAREHAB1', 'USEEQUIP']
new_features = ['has_physical_activity', 'exercised_past_30d', 'primary_exercise_duration', 
                'secondary_exercise_duration', 'meets_strength_training', 'primary_exercise_freq', 
                'secondary_exercise_freq', 'has_strength_training', 'strength_training_freq', 
                'total_exercise_freq', 'primary_activity_duration', 'secondary_activity_duration', 
                'total_activity_duration', 'vo2_max', 'in_strength_rehab', 'in_heart_rehab', 
                'uses_exercise_equipment', 'fitness_program_participation']

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_9.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load dataset
data_path = '../data/processed/BRFSS_9.csv'

df = pd.read_csv(data_path)

# Define Diet & Nutrition features
diet_nutrition_features = {
    'Fruits_Vegetables': [
        'FRUIT1', 'FRUITJU1', 'FVBEANS', 'FVGREEN', 'FVORANG', 'VEGETAB1',
        '_FRUTSUM', '_VEGESUM', 'FTJUDA1_', 'GRENDAY_', 'VEGEDA1_', 'FRUTDA1_',
        'BEANDAY_', 'ORNGDAY_', '_FRUITEX', '_VEGETEX', '_MISFRTN', '_VEG23', '_FRT16'
    ],
    'Fruit_Veg_Response': ['_FRTRESP', '_VEGRESP', '_MISVEGN']
}

# Flatten and keep only features present in df
all_diet_features = (
    diet_nutrition_features['Fruits_Vegetables'] +
    diet_nutrition_features['Fruit_Veg_Response']
)
all_diet_features = df.columns.intersection(all_diet_features).tolist()

if not all_diet_features:
    raise ValueError("No diet/nutrition features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_diet(series):
    name = series.name
    if name in [
        'FRUIT1', 'FRUITJU1', 'FVBEANS', 'FVGREEN', 'FVORANG', 'VEGETAB1',
        'FTJUDA1_', 'GRENDAY_', 'VEGEDA1_', 'FRUTDA1_', 'BEANDAY_', 'ORNGDAY_',
        '_FRUTSUM', '_VEGESUM'
    ]:
        return series.replace({777: np.nan, 999: np.nan, 300: 0, 555: 0})
    elif name in ['_FRUITEX', '_VEGETEX', '_MISFRTN', '_VEG23', '_FRT16', '_FRTRESP', '_VEGRESP', '_MISVEGN']:
        return series.replace({7: np.nan, 9: np.nan})
    return series

df[all_diet_features] = df[all_diet_features].apply(clean_brfss_diet)

# Remove features that are entirely NaN
valid_features = [col for col in all_diet_features if df[col].notna().any()]
for col in all_diet_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All diet/nutrition features are NaN after cleaning.")

# ======================
# Step 2: Engineer Fruits_Vegetables features
# ======================
# Standardize servings per day (divide by 100 for derived variables)
df['fruit_servings_day'] = df['FRUTDA1_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['juice_servings_day'] = df['FTJUDA1_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['bean_servings_day'] = df['BEANDAY_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['green_veg_servings_day'] = df['GRENDAY_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['orange_veg_servings_day'] = df['ORNGDAY_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['other_veg_servings_day'] = df['VEGEDA1_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)

# Standardize frequency variables to times per week
def standardize_freq(value):
    if pd.isna(value):
        return np.nan
    if 101 <= value <= 199:
        return value - 100  # Weekly
    elif 201 <= value <= 299:
        return (value - 200) / 4.33  # Monthly to weekly
    return 0  # Includes 300=never, 555=less than daily

for col in ['FRUIT1', 'FRUITJU1', 'FVBEANS', 'FVGREEN', 'FVORANG', 'VEGETAB1']:
    if col in df.columns:
        df[f'{col.lower()}_freq_week'] = df[col].apply(standardize_freq)

# Total fruit and vegetable servings
def total_fruit_servings(row):
    servings = [row['fruit_servings_day'], row['juice_servings_day']]
    if any(pd.isna(s) for s in servings):
        return np.nan
    return sum(s for s in servings if not pd.isna(s))

def total_veg_servings(row):
    servings = [row['bean_servings_day'], row['green_veg_servings_day'], 
                row['orange_veg_servings_day'], row['other_veg_servings_day']]
    if any(pd.isna(s) for s in servings):
        return np.nan
    return sum(s for s in servings if not pd.isna(s))

df['total_fruit_servings'] = df.apply(total_fruit_servings, axis=1)
df['total_veg_servings'] = df.apply(total_veg_servings, axis=1)
df['total_fv_servings'] = df[['total_fruit_servings', 'total_veg_servings']].sum(axis=1, skipna=True)

# Binary indicators for meeting recommendations
df['meets_fruit_recommendation'] = np.where(df['_FRT16'] == 1, 1, 
                                           np.where(df['_FRT16'] == 0, 0, np.nan))
df['meets_veg_recommendation'] = np.where(df['_VEG23'] == 1, 1, 
                                          np.where(df['_VEG23'] == 0, 0, np.nan))

# Composite: Fruit and vegetable recommendation score
def fv_recommendation_score(row):
    conditions = [row['meets_fruit_recommendation'], row['meets_veg_recommendation']]
    if any(pd.isna(c) for c in conditions):
        return np.nan
    return sum(c == 1 for c in conditions)

df['fv_recommendation_score'] = df.apply(fv_recommendation_score, axis=1)

# ======================
# Step 3: Engineer Fruit_Veg_Response features
# ======================
df['fruit_response_complete'] = np.where(df['_FRTRESP'] == 1, 1, 
                                        np.where(df['_FRTRESP'] == 0, 0, np.nan))
df['veg_response_complete'] = np.where(df['_VEGRESP'] == 1, 1, 
                                      np.where(df['_VEGRESP'] == 0, 0, np.nan))
df['veg_data_missing'] = np.where(df['_MISVEGN'] == 0, 0, 
                                  np.where(df['_MISVEGN'] >= 1, 1, np.nan))

# Updated diet_nutrition_features dictionary
diet_nutrition_features = {
    'Fruits_Vegetables': [
        'FRUIT1', 'FRUITJU1', 'FVBEANS', 'FVGREEN', 'FVORANG', 'VEGETAB1',
        '_FRUTSUM', '_VEGESUM', 'FTJUDA1_', 'GRENDAY_', 'VEGEDA1_', 'FRUTDA1_',
        'BEANDAY_', 'ORNGDAY_', '_FRUITEX', '_VEGETEX', '_MISFRTN', '_VEG23', '_FRT16',
        'fruit_servings_day', 'juice_servings_day', 'bean_servings_day', 
        'green_veg_servings_day', 'orange_veg_servings_day', 'other_veg_servings_day',
        'fruit1_freq_week', 'fruitju1_freq_week', 'fvbeans_freq_week', 
        'fvgreen_freq_week', 'fvorang_freq_week', 'vegetab1_freq_week',
        'total_fruit_servings', 'total_veg_servings', 'total_fv_servings',
        'meets_fruit_recommendation', 'meets_veg_recommendation', 'fv_recommendation_score'
    ],
    'Fruit_Veg_Response': [
        '_FRTRESP', '_VEGRESP', '_MISVEGN',
        'fruit_response_complete', 'veg_response_complete', 'veg_data_missing'
    ]
}

# ======================
# Step 4: Display value counts for categorical/binary features
# ======================
categorical_cols = [
    'meets_fruit_recommendation', 'meets_veg_recommendation', 'fv_recommendation_score',
    'fruit_response_complete', 'veg_response_complete', 'veg_data_missing'
]

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 5: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Numeric Features: Boxplots and Histograms
numeric_cols = [
    'fruit_servings_day', 'juice_servings_day', 'bean_servings_day', 
    'green_veg_servings_day', 'orange_veg_servings_day', 'other_veg_servings_day',
    'fruit1_freq_week', 'fruitju1_freq_week', 'fvbeans_freq_week', 
    'fvgreen_freq_week', 'fvorang_freq_week', 'vegetab1_freq_week',
    'total_fruit_servings', 'total_veg_servings', 'total_fv_servings',
    'fv_recommendation_score'
]
valid_numeric_cols = [col for col in numeric_cols if col in df.columns]
if valid_numeric_cols:
    n_cols = 3
    n_rows = (len(valid_numeric_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(valid_numeric_cols):
        sns.boxplot(y=df[col], ax=axes[i])
        axes[i].set_title(f'Boxplot of {col}')
        axes[i].set_ylabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, col in enumerate(valid_numeric_cols):
        sns.histplot(df[col].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {col}')
        axes[i].set_xlabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

# Categorical Features: Barplots
if categorical_cols:
    n_cols = 3
    n_rows = (len(categorical_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(categorical_cols):
        if col in df.columns:
            order = df[col].value_counts(dropna=False).index
            sns.countplot(x=df[col], order=order, ax=axes[i])
            axes[i].set_title(f'Bar Plot of {col}')
            axes[i].set_xlabel(col)
            axes[i].tick_params(axis='x', rotation=45)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

# ======================
# Step 6: Save updated dataset
# ======================
all_original_features = [
    'FRUIT1', 'FRUITJU1', 'FVBEANS', 'FVGREEN', 'FVORANG', 'VEGETAB1',
    '_FRUTSUM', '_VEGESUM', 'FTJUDA1_', 'GRENDAY_', 'VEGEDA1_', 'FRUTDA1_',
    'BEANDAY_', 'ORNGDAY_', '_FRUITEX', '_VEGETEX', '_MISFRTN', '_VEG23', '_FRT16',
    '_FRTRESP', '_VEGRESP', '_MISVEGN'
]
new_features = [
    'fruit_servings_day', 'juice_servings_day', 'bean_servings_day', 
    'green_veg_servings_day', 'orange_veg_servings_day', 'other_veg_servings_day',
    'fruit1_freq_week', 'fruitju1_freq_week', 'fvbeans_freq_week', 
    'fvgreen_freq_week', 'fvorang_freq_week', 'vegetab1_freq_week',
    'total_fruit_servings', 'total_veg_servings', 'total_fv_servings',
    'meets_fruit_recommendation', 'meets_veg_recommendation', 'fv_recommendation_score',
    'fruit_response_complete', 'veg_response_complete', 'veg_data_missing'
]

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_10.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data_path = '../data/processed/BRFSS_10.csv'
df = pd.read_csv(data_path)

# Define Smoking & Alcohol features
behavior_features = {
    'Smoking_Tobacco': {
        'Standard': ['SMOKE100', 'SMOKDAY2', '_SMOKER3', '_RFSMOK3'],
        'Quit_Attempts': ['STOPSMK2', 'LASTSMK2']
    },
    'Alcohol': {
        'Standard': ['ALCDAY5', 'AVEDRNK2', 'DRNK3GE5', 'MAXDRNKS', '_RFBING5', 'DROCDY3_', 'DRNKANY5']
    }
}

# Flatten features
all_behavior_features = (
    behavior_features['Smoking_Tobacco']['Standard']
    + behavior_features['Smoking_Tobacco']['Quit_Attempts']
    + behavior_features['Alcohol']['Standard']
)
all_behavior_features = df.columns.intersection(all_behavior_features).tolist()

if not all_behavior_features:
    raise ValueError("No smoking/alcohol features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_behavior(series):
    name = series.name
    if name in ['SMOKE100', 'SMOKDAY2', '_SMOKER3', '_RFSMOK3', 'STOPSMK2', 'DRNKANY5', '_RFBING5']:
        return series.replace({7: np.nan, 9: np.nan})
    elif name == 'LASTSMK2':
        return series.replace({77: np.nan, 99: np.nan})
    elif name in ['AVEDRNK2', 'DRNK3GE5', 'MAXDRNKS']:
        return series.replace({77: np.nan, 99: np.nan, 88: 0})
    elif name == 'ALCDAY5':
        return series.replace({777: np.nan, 999: np.nan, 888: 0})
    elif name == 'DROCDY3_':
        return series.replace({7777: np.nan, 9999: np.nan})
    return series

df[all_behavior_features] = df[all_behavior_features].apply(clean_brfss_behavior)

# Remove all-NaN features
valid_features = [col for col in all_behavior_features if df[col].notna().any()]
for col in all_behavior_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All smoking/alcohol features are NaN after cleaning.")

# ======================
# Step 2: Engineer Smoking_Tobacco features
# ======================
# Standard
df['ever_smoked'] = np.where(df['SMOKE100'] == 1, 1, 
                            np.where(df['SMOKE100'] == 2, 0, np.nan))
df['current_smoker'] = np.where(df['_RFSMOK3'] == 2, 1, 
                               np.where(df['_RFSMOK3'] == 1, 0, np.nan))
df['smoking_frequency'] = df['SMOKDAY2'].replace({1: 2, 2: 1, 3: 0, np.nan: np.nan})  # 0=Not at all, 1=Some days, 2=Every day
df['smoking_status'] = df['_SMOKER3']  # 1=Current every day, 2=Current some days, 3=Former, 4=Never

# Quit Attempts
df['tried_quit_smoking'] = np.where(df['STOPSMK2'] == 1, 1, 
                                   np.where(df['STOPSMK2'] == 2, 0, np.nan))
df['years_since_quit'] = df['LASTSMK2'].map({1: 0.1, 2: 0.2, 3: 0.5, 4: 1.5, 5: 4, 6: 7.5, 7: 10, 8: 0, np.nan: np.nan})

# Composite: Smoking risk score
def smoking_risk_score(row):
    if pd.isna(row['smoking_status']):
        return np.nan
    if row['smoking_status'] == 4:
        return 0  # Never
    elif row['smoking_status'] == 3:
        return 1  # Former
    elif row['smoking_status'] == 2:
        return 2  # Current some days
    elif row['smoking_status'] == 1:
        return 3  # Current every day
    return np.nan

df['smoking_risk_score'] = df.apply(smoking_risk_score, axis=1)

# ======================
# Step 3: Engineer Alcohol features
# ======================
df['any_alcohol'] = np.where(df['DRNKANY5'] == 1, 1, 
                            np.where(df['DRNKANY5'] == 2, 0, np.nan))
df['binge_drinker'] = np.where(df['_RFBING5'] == 2, 1, 
                              np.where(df['_RFBING5'] == 1, 0, np.nan))
df['avg_drinks_per_day'] = df['AVEDRNK2']
df['max_drinks_occasion'] = df['MAXDRNKS']
df['drinks_per_day'] = df['DROCDY3_'].apply(lambda x: x / 100 if not pd.isna(x) else np.nan)
df['binge_days'] = df['DRNK3GE5']

# Standardize drinking days to times per week
def standardize_drinking_days(value):
    if pd.isna(value):
        return np.nan
    if value == 0:
        return 0
    if 101 <= value <= 199:
        return value - 100  # Weekly
    elif 201 <= value <= 299:
        return (value - 200) / 4.33  # Monthly to weekly
    return 0

df['drinking_days_week'] = df['ALCDAY5'].apply(standardize_drinking_days)

# Composite: Alcohol risk score
def alcohol_risk_score(row):
    conditions = [
        row['any_alcohol'],
        row['binge_drinker'],
        1 if not pd.isna(row['drinks_per_day']) and row['drinks_per_day'] > 2 else 0
    ]
    if any(pd.isna(c) for c in conditions[:2]):
        return np.nan
    return sum(c == 1 for c in conditions)

df['alcohol_risk_score'] = df.apply(alcohol_risk_score, axis=1)

# Updated behavior_features dictionary
behavior_features = {
    'Smoking_Tobacco': {
        'Standard': ['SMOKE100', 'SMOKDAY2', '_SMOKER3', '_RFSMOK3', 
                     'ever_smoked', 'current_smoker', 'smoking_frequency', 'smoking_status', 'smoking_risk_score'],
        'Quit_Attempts': ['STOPSMK2', 'LASTSMK2', 'tried_quit_smoking', 'years_since_quit']
    },
    'Alcohol': {
        'Standard': ['ALCDAY5', 'AVEDRNK2', 'DRNK3GE5', 'MAXDRNKS', '_RFBING5', 'DROCDY3_', 'DRNKANY5',
                     'any_alcohol', 'binge_drinker', 'drinking_days_week', 'avg_drinks_per_day', 
                     'max_drinks_occasion', 'drinks_per_day', 'binge_days', 'alcohol_risk_score']
    }
}

# ======================
# Step 4: Display value counts for categorical/binary features
# ======================
categorical_cols = [
    'ever_smoked', 'current_smoker', 'smoking_frequency', 'smoking_status', 
    'tried_quit_smoking', 'smoking_risk_score', 'any_alcohol', 'binge_drinker', 
    'alcohol_risk_score'
]

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 5: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Numeric Features: Boxplots and Histograms
numeric_cols = [
    'years_since_quit', 'drinking_days_week', 'avg_drinks_per_day', 
    'max_drinks_occasion', 'drinks_per_day', 'binge_days', 
    'smoking_risk_score', 'alcohol_risk_score'
]
valid_numeric_cols = [col for col in numeric_cols if col in df.columns]
if valid_numeric_cols:
    n_cols = 3
    n_rows = (len(valid_numeric_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(valid_numeric_cols):
        sns.boxplot(y=df[col], ax=axes[i])
        axes[i].set_title(f'Boxplot of {col}')
        axes[i].set_ylabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, col in enumerate(valid_numeric_cols):
        sns.histplot(df[col].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {col}')
        axes[i].set_xlabel(col)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

# Categorical Features: Barplots
if categorical_cols:
    n_cols = 3
    n_rows = (len(categorical_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(categorical_cols):
        if col in df.columns:
            order = df[col].value_counts(dropna=False).index
            sns.countplot(x=df[col], order=order, ax=axes[i])
            axes[i].set_title(f'Bar Plot of {col}')
            axes[i].set_xlabel(col)
            axes[i].tick_params(axis='x', rotation=45)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()


# ======================
# Step 6: Save updated dataset
# ======================
all_original_features = [
    'SMOKE100', 'SMOKDAY2', '_SMOKER3', '_RFSMOK3', 'STOPSMK2', 'LASTSMK2',
    'ALCDAY5', 'AVEDRNK2', 'DRNK3GE5', 'MAXDRNKS', '_RFBING5', 'DROCDY3_', 'DRNKANY5'
]
new_features = [
    'ever_smoked', 'current_smoker', 'smoking_frequency', 'smoking_status', 
    'tried_quit_smoking', 'years_since_quit', 'smoking_risk_score',
    'any_alcohol', 'binge_drinker', 'drinking_days_week', 'avg_drinks_per_day', 
    'max_drinks_occasion', 'drinks_per_day', 'binge_days', 'alcohol_risk_score'
]

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_11.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Load dataset
data_path = '../data/processed/BRFSS_12.csv'


df = pd.read_csv(data_path)

# Define Income, Education, Demographics features
socio_demo_features = {
    'Income_Education': {
        'Income': ['INCOME2', '_INCOMG'],
        'Education': ['EDUCA', '_EDUCAG']
    },
    'Demographics': {
        'Basic': ['SEX', 'RCSRLTN2', 'MARITAL', 'CHILDREN',
                  'NUMWOMEN', 'NUMMEN', 'NUMADULT', 'CADULT',
                  'RCSGENDR', '_CHLDCNT']
    }
}

# Flatten & keep only features present in dataset
all_socio_features = (
    socio_demo_features['Income_Education']['Income']
    + socio_demo_features['Income_Education']['Education']
    + socio_demo_features['Demographics']['Basic']
)
all_socio_features = df.columns.intersection(all_socio_features).tolist()

if not all_socio_features:
    raise ValueError("No socioeconomic/demographic features found in dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_socio(series):
    name = series.name
    # Income: 77/99=missing (INCOME2), 9=missing (_INCOMG)
    if name == 'INCOME2':
        return series.replace({77: np.nan, 99: np.nan})
    elif name == '_INCOMG':
        return series.replace({9: np.nan})
    # Education: 9=missing
    elif name in ['EDUCA', '_EDUCAG']:
        return series.replace({9: np.nan})
    # Demographics categorical: 9=missing
    elif name in ['SEX', 'MARITAL', 'RCSRLTN2', 'RCSGENDR']:
        return series.replace({9: np.nan})
    # Children counts: 88=None, 99=Refused
    elif name in ['CHILDREN', 'NUMWOMEN', 'NUMMEN', 'NUMADULT', '_CHLDCNT']:
        return series.replace({88: 0, 99: np.nan})
    # CADULT: No specific codes, keep NaNs
    return series

df[all_socio_features] = df[all_socio_features].apply(clean_brfss_socio)

# Remove all-NaN features
valid_features = [col for col in all_socio_features if df[col].notna().any()]
for col in all_socio_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All socioeconomic/demographic features are NaN after cleaning.")

# ======================
# Step 2: Engineer Income features
# ======================
df['income_level'] = df['_INCOMG']  # 1=<15k, ..., 5=50k+
df['high_income'] = np.where(df['_INCOMG'] >= 5, 1, 
                            np.where(df['_INCOMG'].notna(), 0, np.nan))

# Composite: Income disparity (map INCOME2 to _INCOMG scale, normalize difference)
def income_disparity(row):
    income2 = row['INCOME2']
    incomg = row['_INCOMG']
    if pd.isna(income2) or pd.isna(incomg):
        return np.nan
    # Map INCOME2 (1-8) to _INCOMG (1-5) roughly
    income2_mapped = {
        1: 1, 2: 1,  # <15k
        3: 2, 4: 2,  # 15k-25k
        5: 3,  # 25k-35k
        6: 4,  # 35k-50k
        7: 5, 8: 5   # 50k+
    }.get(income2, np.nan)
    if pd.isna(income2_mapped):
        return np.nan
    return abs(income2_mapped - incomg) / 4  # Normalize to 0-1

df['income_disparity'] = df.apply(income_disparity, axis=1)

# ======================
# Step 3: Engineer Education features
# ======================
df['education_level'] = df['_EDUCAG']  # 1=≤Grade 12, ..., 4=College 4+
df['college_educated'] = np.where(df['_EDUCAG'] == 4, 1, 
                                 np.where(df['_EDUCAG'].notna(), 0, np.nan))

# Composite: Education consistency
def edu_consistency(row):
    educa = row['EDUCA']
    educag = row['_EDUCAG']
    if pd.isna(educa) or pd.isna(educag):
        return np.nan
    # Map EDUCA (1-6) to _EDUCAG (1-4)
    educa_mapped = {
        1: 1, 2: 1, 3: 1,  # ≤Grade 12
        4: 2,  # High school
        5: 3,  # Some college
        6: 4   # College 4+
    }.get(educa, np.nan)
    if pd.isna(educa_mapped):
        return np.nan
    return 1 if educa_mapped == educag else 0

df['edu_consistency'] = df.apply(edu_consistency, axis=1)

# ======================
# Step 4: Engineer Demographics features
# ======================
df['is_female'] = np.where(df['SEX'] == 2, 1, 
                          np.where(df['SEX'] == 1, 0, np.nan))
df['is_married'] = np.where(df['MARITAL'] == 1, 1, 
                           np.where(df['MARITAL'].notna(), 0, np.nan))
df['has_partner'] = np.where(df['RCSRLTN2'].isin([1, 2]), 1, 
                            np.where(df['RCSRLTN2'].notna(), 0, np.nan))
df['num_children'] = df['CHILDREN']
df['num_women'] = df['NUMWOMEN']
df['num_men'] = df['NUMMEN']
df['num_adults'] = df['NUMADULT']
df['imputed_children'] = df['_CHLDCNT']
df['partner_gender'] = df['RCSGENDR']  # 1=Male, 2=Female

# Composite: Household size
def household_size(row):
    components = [row['num_women'], row['num_men'], row['num_children']]
    if any(pd.isna(c) for c in components):
        return np.nan
    return sum(components)

df['household_size'] = df.apply(household_size, axis=1)

# Cap high-variance numeric features
df['num_children'] = df['num_children'].clip(upper=10)
df['num_women'] = df['num_women'].clip(upper=10)
df['num_men'] = df['num_men'].clip(upper=10)
df['num_adults'] = df['num_adults'].clip(upper=10)
df['household_size'] = df['household_size'].clip(upper=20)

# Skip CADULT due to high NaNs and redundancy with NUMADULT
if 'CADULT' in df.columns and df['CADULT'].notna().sum() == 0:
    print("Warning: CADULT is entirely NaN and will be skipped.")

# ======================
# Step 5: Display value counts for categorical/binary features
# ======================
categorical_cols = [
    'income_level', 'high_income', 'education_level', 'college_educated',
    'edu_consistency', 'is_female', 'is_married', 'has_partner',
    'imputed_children', 'partner_gender'
]

for col in categorical_cols:
    if col in df.columns:
        vc = df[col].value_counts(dropna=False).sort_index()
        vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
        vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
        print(f"\nValue Counts for {col}:")
        print(vc_df)

# ======================
# Step 6: Visualization
# ======================
plt.style.use('seaborn-v0_8')

# Numeric Features: Boxplots and Histograms
numeric_cols = ['num_children', 'num_women', 'num_men', 'num_adults', 'household_size']
valid_numeric_cols = [col for col in numeric_cols if col in df.columns and df[col].notna().sum() > 0]
if valid_numeric_cols:
    n_cols = 3
    n_rows = (len(valid_numeric_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    i = 0
    for col in valid_numeric_cols:
        if df[col].notna().sum() > 0:
            try:
                sns.boxplot(y=df[col], ax=axes[i])
                axes[i].set_title(f'Boxplot of {col}')
                axes[i].set_ylabel(col)
                i += 1
            except Exception as e:
                print(f"Warning: Could not plot boxplot for {col}. Error: {str(e)}")
        else:
            print(f"Warning: Skipping boxplot for {col} due to no valid data.")
    for j in range(i, len(axes)):
        axes[j].set_visible(False)
    if i > 0:
        plt.tight_layout()
        plt.show()
    else:
        print("\nNo valid numeric boxplots to display.")

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    i = 0
    for col in valid_numeric_cols:
        if df[col].notna().sum() > 0:
            try:
                sns.histplot(df[col].dropna(), bins=20, kde=True, ax=axes[i])
                axes[i].set_title(f'Histogram of {col}')
                axes[i].set_xlabel(col)
                i += 1
            except Exception as e:
                print(f"Warning: Could not plot histogram for {col}. Error: {str(e)}")
        else:
            print(f"Warning: Skipping histogram for {col} due to no valid data.")
    for j in range(i, len(axes)):
        axes[j].set_visible(False)
    if i > 0:
        plt.tight_layout()
        plt.show()
    else:
        print("\nNo valid numeric histograms to display.")

# Categorical Features: Barplots
valid_categorical_cols = [col for col in categorical_cols if col in df.columns and df[col].notna().sum() > 0]
if valid_categorical_cols:
    n_cols = 3
    n_rows = (len(valid_categorical_cols) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(valid_categorical_cols):
        if df[col].notna().sum() > 0:
            order = df[col].value_counts(dropna=False).index
            sns.countplot(x=df[col], order=order, ax=axes[i])
            axes[i].set_title(f'Bar Plot of {col}')
            axes[i].set_xlabel(col)
            axes[i].tick_params(axis='x', rotation=45)
        else:
            print(f"Warning: Skipping bar plot for {col} due to no valid data.")
            axes[i].set_visible(False)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()



# ======================
# Step 7: Save updated dataset
# ======================
all_original_features = [
    'INCOME2', '_INCOMG', 'EDUCA', '_EDUCAG', 'SEX', 'RCSRLTN2', 'MARITAL',
    'CHILDREN', 'NUMWOMEN', 'NUMMEN', 'NUMADULT', 'CADULT', 'RCSGENDR', '_CHLDCNT'
]
new_features = [
    'income_level', 'high_income', 'income_disparity', 'education_level',
    'college_educated', 'edu_consistency', 'is_female', 'is_married',
    'has_partner', 'num_children', 'num_women', 'num_men', 'num_adults',
    'imputed_children', 'partner_gender', 'household_size'
]

df = df.drop(columns=[f for f in all_original_features if f in df.columns], errors='ignore')
updated_path = '../data/processed/BRFSS_13.csv'
df.to_csv(updated_path, index=False)
print(f"\nUpdated dataset saved to: {updated_path}")
```
