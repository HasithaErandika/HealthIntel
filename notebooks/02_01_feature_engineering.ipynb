```python
# Create age_cat (CORRECTED & VECTORIZED)

# Initialize with NaN
df['age_cat'] = np.nan

# Priority 1: Use _AGE_G (1-6 are valid)
valid_age_g = df['_AGE_G'].isin([1, 2, 3, 4, 5, 6])
df.loc[valid_age_g, 'age_cat'] = df.loc[valid_age_g, '_AGE_G']

# Use _AGEG5YR 
remaining = df['age_cat'].isna()
ageg_map = {
    1: 1,           # 18-24
    2: 2, 3: 2,     # 25-34
    4: 3, 5: 3,     # 35-44
    6: 4, 7: 4,     # 45-54
    8: 5, 9: 5,     # 55-64
    10: 6, 11: 6, 12: 6, 13: 6, 14: 6  # 65+
}
df.loc[remaining, 'age_cat'] = df.loc[remaining, '_AGEG5YR'].map(ageg_map)

# Use _AGE65YR ONLY to identify 65+ (category 6)
remaining = df['age_cat'].isna()
# Only map confirmed 65+ cases
df.loc[remaining & (df['_AGE65YR'] == 1), 'age_cat'] = 6

# Ensure integer type 
df['age_cat'] = df['age_cat'].astype('Int64')  # nullable integer

# ======================
# Step 2: Create asthma_age_cat
# ======================
def create_asthma_age_cat(age):
    if pd.isna(age) or age in [97, 98, 99]:  # 98 often = "Don't know"
        return 'Missing'
    elif age < 30:
        return 'Early'
    elif 30 <= age <= 49:
        return 'Adult'
    elif 50 <= age <= 64:
        return 'Late_Adult'
    else:  # age >= 65
        return 'Senior'

df['asthma_age_cat'] = df['ASTHMAGE'].apply(create_asthma_age_cat)

# ======================
# Step 3: Create diabetes_age_cat
# ======================
def create_diabetes_age_cat(age):
    if pd.isna(age) or age in [77, 98, 99]:  # 77/98/99 = missing in BRFSS
        return 'Missing'
    elif age < 30:
        return 'Early'
    elif 30 <= age <= 49:
        return 'Adult'
    elif 50 <= age <= 64:
        return 'Late_Adult'
    else:
        return 'Senior'

df['diabetes_age_cat'] = df['DIABAGE2'].apply(create_diabetes_age_cat)

# ======================
# Step 4: Create household_size_cat
# ======================
def create_household_size_cat(size):
    if pd.isna(size) or size in [7, 8, 9]:  # 7/8/9 often = "Refused"/"Don't know"/"9+"
        return 'Missing'
    elif size == 1:
        return 'Single'
    elif size == 2:
        return 'Standard'
    else:  # 3-6 (or 3-8 if 9 is excluded)
        return 'Large'

df['household_size_cat'] = df['NUMHHOL2'].apply(create_household_size_cat)

# ======================
# Step 5: Display value counts
# ======================
categorical_cols = ['age_cat', 'asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']

for col in categorical_cols:
    vc = df[col].value_counts(dropna=False).sort_index()
    vc_percent = df[col].value_counts(dropna=False, normalize=True).sort_index() * 100
    vc_df = pd.DataFrame({'Count': vc, 'Percentage': vc_percent})
    print(f"\nValue Counts for {col}:")
    print(vc_df)

# ======================
# Step 6: Visualization (FIXED LAYOUT)
# ======================
plt.style.use('seaborn-v0_8')
fig = plt.figure(figsize=(20, 16))

# Bar plots (2x2 grid for 4 plots)
bar_cols = ['age_cat', 'asthma_age_cat', 'diabetes_age_cat', 'household_size_cat']
for i, col in enumerate(bar_cols):
    ax = plt.subplot(2, 2, i + 1)
    # Sort by count for better readability
    order = df[col].value_counts().index
    sns.countplot(data=df, x=col, ax=ax, order=order)
    ax.set_title(f'Distribution of {col}', fontsize=14)
    ax.set_xlabel(col, fontsize=12)
    ax.set_ylabel('Count', fontsize=12)
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Box plots 
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Asthma age by current age group
sns.boxplot(data=df[df['asthma_age_cat'] != 'Missing'], 
            x='age_cat', y='ASTHMAGE', ax=axes[0])
axes[0].set_title('Asthma Onset Age by Current Age Group')
axes[0].set_xlabel('Current Age Group (age_cat)')
axes[0].set_ylabel('Asthma Onset Age')

# Diabetes age by current age group
sns.boxplot(data=df[df['diabetes_age_cat'] != 'Missing'], 
            x='age_cat', y='DIABAGE2', ax=axes[1])
axes[1].set_title('Diabetes Onset Age by Current Age Group')
axes[1].set_xlabel('Current Age Group (age_cat)')
axes[1].set_ylabel('Diabetes Onset Age')

plt.tight_layout()
plt.show()
```

```python

# Load the data
data_path = '../data/processed/BRFSS_3.csv'
df = pd.read_csv(data_path)

# Define general health features
general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3'],
    'Quality_of_Life': ['QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'PAFREQ1_', 'PAFREQ2_', 'QLSTRES2'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH']
}

# Flatten and keep only features present in df
all_general_health_features = (
    general_health_features['Standard']
    + general_health_features['Quality_of_Life']
    + general_health_features['Mental_Health']
)
all_general_health_features = df.columns.intersection(all_general_health_features).tolist()

if not all_general_health_features:
    raise ValueError("No general health features found in the dataset.")

# BRFSS-specific cleaning function
def clean_brfss_values(series):
    name = series.name
    if name in ['PHYSHLTH', 'MENTHLTH', 'POORHLTH']:
        # Days: 1-30 valid, 77=Don't know, 88=None, 99=Refused
        series = series.replace({77: np.nan, 99: np.nan, 88: 0})
    elif name in ['GENHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3',
                  'QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'QLSTRES2', 'HADSIGM3']:
        # Categorical: 7/9 = missing
        series = series.replace({7: np.nan, 9: np.nan})
    elif name in ['PAFREQ1_', 'PAFREQ2_']:
        # Physical activity: 99999 = refused
        series = series.replace({99999: np.nan})
    return series

# Apply cleaning
df[all_general_health_features] = df[all_general_health_features].apply(clean_brfss_values)

# Remove features that are ALL NaN after cleaning
valid_features = []
for col in all_general_health_features:
    if df[col].notna().any():
        valid_features.append(col)
    else:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All general health features are NaN after cleaning.")

# Update feature lists
numeric_features = [f for f in valid_features if pd.api.types.is_numeric_dtype(df[f])]
categorical_features = [f for f in valid_features if f not in numeric_features]

# -------------------------------
# Summary Statistics (unchanged)
# -------------------------------
print("Summary statistics for all general health features:\n")
for feature in valid_features:
    print(f"\nFeature: {feature}")
    if pd.api.types.is_numeric_dtype(df[feature]):
        print(df[feature].describe())
    else:
        counts = df[feature].value_counts(dropna=False)
        percentages = (counts / len(df) * 100).round(2)
        print(pd.DataFrame({'Count': counts, 'Percentage': percentages}))

# -------------------------------
# Visualization with 3x3 grids
# -------------------------------
plt.style.use('seaborn-v0_8')

# --- Numeric Features: Boxplots in 3x3 grids ---
if numeric_features:
    n_numeric = len(numeric_features)
    n_cols = 3
    n_rows = (n_numeric + n_cols - 1) // n_cols  # Ceiling division

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_numeric > 1 else [axes]

    for i, feature in enumerate(numeric_features):
        sns.boxplot(y=df[feature], ax=axes[i])
        axes[i].set_title(f'Boxplot of {feature}')
        axes[i].set_ylabel(feature)

    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()

    # --- Histograms in separate 3x3 grids ---
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_numeric > 1 else [axes]

    for i, feature in enumerate(numeric_features):
        sns.histplot(df[feature].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {feature}')
        axes[i].set_xlabel(feature)

    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()

# --- Categorical Features: Barplots in 3x3 grids ---
if categorical_features:
    n_cat = len(categorical_features)
    n_cols = 3
    n_rows = (n_cat + n_cols - 1) // n_cols

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten() if n_cat > 1 else [axes]

    for i, feature in enumerate(categorical_features):
        order = df[feature].value_counts(dropna=False).index
        sns.countplot(data=df, x=feature, order=order, ax=axes[i])
        axes[i].set_title(f'Bar Plot of {feature}')
        axes[i].set_xlabel(feature)
        axes[i].tick_params(axis='x', rotation=45)

    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
# Load the data
data_path = '../data/processed/BRFSS_3.csv'
df = pd.read_csv(data_path)

# Define general health features
general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3'],
    'Quality_of_Life': ['QLMENTL2', 'QLHLTH2', '_RFHLTH', 'PAINACT2', 'PAFREQ1_', 'PAFREQ2_', 'QLSTRES2'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH']
}

# Flatten and filter valid features
all_general_health_features = (
    general_health_features['Standard']
    + general_health_features['Quality_of_Life']
    + general_health_features['Mental_Health']
)
all_general_health_features = df.columns.intersection(all_general_health_features).tolist()

if not all_general_health_features:
    raise ValueError("No general health features found in the dataset.")

# ======================
# Step 1: Clean BRFSS special codes
# ======================
def clean_brfss_values(series):
    if series.name in ['PHYSHLTH', 'MENTHLTH', 'POORHLTH']:
        return series.where(~series.isin([77, 99]), np.nan).replace(88, 0)
    elif series.name in ['GENHLTH']:
        return series.where(~series.isin([7, 9]), np.nan)
    elif series.name in ['QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3', 'HADSIGM3', '_RFHLTH']:
        return series.where(~series.isin([7, 9]), np.nan)
    elif series.name in ['PAFREQ1_', 'PAFREQ2_']:
        return series.where(~series.isin([99999, 99000]), np.nan)  # Include 99000
    return series

df[all_general_health_features] = df[all_general_health_features].apply(clean_brfss_values)

# ======================
# Step 2: Engineer Standard features
# ======================
df['general_health_cat'] = np.where(df['GENHLTH'].isin([1, 2, 3]), 1,
                                   np.where(df['GENHLTH'].isin([4, 5]), 0, np.nan))

df['physical_health_bad'] = np.where(df['PHYSHLTH'] >= 1, 1,
                                    np.where(df['PHYSHLTH'] == 0, 0, np.nan))

df['poor_health_days'] = np.where(df['POORHLTH'] >= 1, 1,
                                  np.where(df['POORHLTH'] == 0, 0, np.nan))

def binarize_brfss(col):
    return np.where(col == 1, 1, np.where(col == 2, 0, np.nan))

df['has_activity_limitation'] = binarize_brfss(df['QLACTLM2'])
df['has_depression'] = binarize_brfss(df['ADDEPEV2'])
df['has_asthma'] = binarize_brfss(df['ASTHMA3'])
df['has_arthritis'] = binarize_brfss(df['HAVARTH3'])

# ======================
# Step 3: Engineer Quality_of_Life features
# ======================
df['general_health_risk'] = binarize_brfss(df['_RFHLTH'])  # 1 = fair/poor health (risk)

def standardize_activity_freq(value):
    if pd.isna(value) or value in [99000, 99999]:
        return np.nan
    if 1000 <= value <= 1999:
        return float(value - 1000)
    elif 2000 <= value <= 2999:
        return float((value - 2000) / 4.33)
    return np.nan

df['vigorous_activity_freq'] = df['PAFREQ1_'].apply(standardize_activity_freq)
df['moderate_activity_freq'] = df['PAFREQ2_'].apply(standardize_activity_freq)

# Engineer Mental_Health features (VECTORIZED)

df['has_anxiety'] = binarize_brfss(df['HADSIGM3'])
df['mental_health_bad'] = np.where(df['MENTHLTH'] >= 1, 1,
                                  np.where(df['MENTHLTH'] == 0, 0, np.nan))

# Vectorized composite: anxiety OR ≥1 mental bad day
anxiety = df['has_anxiety']
mental_bad = df['mental_health_bad']
df['mental_health_condition'] = np.nan
df.loc[(anxiety == 1) | (mental_bad == 1), 'mental_health_condition'] = 1
df.loc[(anxiety == 0) & (mental_bad == 0), 'mental_health_condition'] = 0

# Updated feature groups

general_health_features = {
    'Standard': ['GENHLTH', 'PHYSHLTH', 'POORHLTH', 'QLACTLM2', 'ADDEPEV2', 'ASTHMA3', 'HAVARTH3',
                 'general_health_cat', 'physical_health_bad', 'poor_health_days', 'has_activity_limitation',
                 'has_depression', 'has_asthma', 'has_arthritis'],
    'Quality_of_Life': ['_RFHLTH', 'PAFREQ1_', 'PAFREQ2_', 'general_health_risk', 
                        'vigorous_activity_freq', 'moderate_activity_freq'],
    'Mental_Health': ['HADSIGM3', 'MENTHLTH', 'has_anxiety', 'mental_health_bad', 'mental_health_condition']
}


categorical_cols = [
    'general_health_cat', 'physical_health_bad', 'poor_health_days', 'has_activity_limitation',
    'has_depression', 'has_asthma', 'has_arthritis',
    'general_health_risk', 'has_anxiety', 'mental_health_condition'
]
categorical_cols = [col for col in categorical_cols if col in df.columns]

numeric_cols = ['vigorous_activity_freq', 'moderate_activity_freq']
numeric_cols = [col for col in numeric_cols if col in df.columns]

# ======================
# DIAGNOSTIC: Print non-null counts
# ======================
print("=== Non-null counts for binary features ===")
for col in categorical_cols:
    total = len(df)
    non_null = df[col].notna().sum()
    print(f"{col:25} | Non-null: {non_null:7,} ({100 * non_null / total:.1f}%)")

print("\n=== Non-null counts for numeric features ===")
for col in numeric_cols:
    total = len(df)
    non_null = df[col].notna().sum()
    print(f"{col:25} | Non-null: {non_null:7,} ({100 * non_null / total:.1f}%)")


# Visualization: Bar plots for binary, Strip plots for numeric

plt.style.use('seaborn-v0_8')

# Plot binary features (bar plots)
if categorical_cols:
    n_cat = len(categorical_cols)
    cols = 3
    rows = (n_cat + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(6 * cols, 4 * rows))
    axes = axes.flatten() if n_cat > 1 else [axes]

    for i, col in enumerate(categorical_cols):
        ax = axes[i]
        # Force categories to 0 and 1
        temp = df[col].dropna()
        if len(temp) == 0:
            ax.text(0.5, 0.5, 'No data', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'{col} (empty)')
        else:
            # Count 0s and 1s explicitly
            counts = temp.value_counts().reindex([0, 1], fill_value=0)
            bars = ax.bar(['No (0)', 'Yes (1)'], counts.values, color=['#1f77b4', '#d62728'])
            ax.set_title(f'{col}\n(Valid: {int(counts.sum())})')
            ax.set_ylabel('Count')
            # Add percentage labels
            total_valid = counts.sum()
            for bar, count in zip(bars, counts.values):
                if count > 0:
                    pct = 100 * count / total_valid
                    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + max(counts) * 0.01,
                            f'{int(count):,}\n({pct:.1f}%)',
                            ha='center', va='bottom')
    
    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
    
    plt.suptitle('Binary Health Indicators', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# Plot numeric features as 1D scatter (strip plots)
if numeric_cols:
    n_num = len(numeric_cols)
    fig, axes = plt.subplots(1, n_num, figsize=(6 * n_num, 5))
    if n_num == 1:
        axes = [axes]
    
    for i, col in enumerate(numeric_cols):
        ax = axes[i]
        data = df[col].dropna()
        if len(data) == 0:
            ax.text(0.5, 0.5, 'No data', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'{col} (empty)')
        else:
            # 1D scatter: jitter x around 0
            x = np.random.normal(0, 0.02, size=len(data))  # small jitter
            ax.scatter(x, data, alpha=0.6, s=10)
            ax.set_title(f'{col} (n={len(data):,})')
            ax.set_ylabel('Frequency (times/week)')
            ax.set_xticks([])  # hide x-axis
            ax.set_xlabel('')
    
    plt.suptitle('Physical Activity Frequency (1D Scatter)', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
```

```python
# Load the data
data_path = '../data/processed/BRFSS_4.csv'
df = pd.read_csv(data_path)

# Define chronic conditions features
chronic_conditions = {
    'Standard': ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'CHKHEMO3'],
    'Diabetes': ['DIABETE3', 'PREDIAB1', 'DOCTDIAB', 'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST'],
    'Hypertension': ['_RFHYPE5', 'BLDSUGAR']
}

# Flatten and keep only features present in df
all_chronic_features = (
    chronic_conditions['Standard']
    + chronic_conditions['Diabetes']
    + chronic_conditions['Hypertension']
)
all_chronic_features = df.columns.intersection(all_chronic_features).tolist()

if not all_chronic_features:
    raise ValueError("No chronic condition features found in the dataset.")

#  Clean BRFSS special codes

def clean_brfss_values(series):
    name = series.name
    if name in ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'DIABETE3', 'PREDIAB1', 
                'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST', '_RFHYPE5']:
        return series.replace({7: np.nan, 9: np.nan})
    elif name in ['CHKHEMO3', 'DOCTDIAB']:
        return series.replace({98: np.nan, 99: np.nan, 88: 0})
    elif name == 'BLDSUGAR':
        return series.replace({999: np.nan})
    return series

df[all_chronic_features] = df[all_chronic_features].apply(clean_brfss_values)

# Remove features that are all NaN after cleaning
valid_features = [col for col in all_chronic_features if df[col].notna().any()]
for col in all_chronic_features:
    if col not in valid_features:
        print(f"Warning: Feature '{col}' is entirely NaN after cleaning and will be skipped.")

if not valid_features:
    raise ValueError("All chronic condition features are NaN after cleaning.")

# Engineer Standard features (CORRECTED LOGIC)

# High BP: 1=Yes on meds, 2=Yes not on meds → both are "has high BP"; 4=borderline → exclude or NaN
df['has_high_bp'] = np.where(df['BPHIGH4'].isin([1, 2]), 1,
                            np.where(df['BPHIGH4'] == 3, 0, np.nan))

df['on_bp_meds'] = np.where(df['BPMEDS'] == 1, 1,
                            np.where(df['BPMEDS'] == 2, 0, np.nan))

df['has_high_cholesterol'] = np.where(df['TOLDHI2'] == 1, 1,
                                     np.where(df['TOLDHI2'] == 2, 0, np.nan))

df['has_kidney_disease'] = np.where(df['CHCKIDNY'] == 1, 1,
                                    np.where(df['CHCKIDNY'] == 2, 0, np.nan))

df['a1c_checks'] = np.where(df['CHKHEMO3'] >= 1, 1,
                            np.where(df['CHKHEMO3'] == 0, 0, np.nan))

# chronic_conditions_standard
std_cols = ['has_high_bp', 'has_high_cholesterol', 'has_kidney_disease']
df['chronic_conditions_standard'] = df[std_cols].sum(axis=1)
df['chronic_conditions_standard'] = df['chronic_conditions_standard'].where(
    df[std_cols].notna().all(axis=1)
)

#  Engineer Diabetes features (CORRECTED LOGIC)

# Diabetes: only DIABETE3 == 1 → Yes; 3 → No; 4 (prediabetes) → should NOT be "No"
df['has_diabetes'] = np.where(df['DIABETE3'] == 1, 1,
                             np.where(df['DIABETE3'] == 3, 0, np.nan))

# Prediabetes: 1=Yes; 2=No; 3=Never tested → NaN
df['has_prediabetes'] = np.where(df['PREDIAB1'] == 1, 1,
                                np.where(df['PREDIAB1'] == 2, 0, np.nan))

df['diabetes_doctor_visits'] = np.where(df['DOCTDIAB'] >= 1, 1,
                                       np.where(df['DOCTDIAB'] == 0, 0, np.nan))

df['diabetes_eye_exam'] = np.where(df['DIABEYE'] == 1, 1,
                                  np.where(df['DIABEYE'] == 2, 0, np.nan))

df['uses_insulin'] = np.where(df['INSULIN'] == 1, 1,
                             np.where(df['INSULIN'] == 2, 0, np.nan))

df['diabetes_education'] = np.where(df['DIABEDU'] == 1, 1,
                                   np.where(df['DIABEDU'] == 2, 0, np.nan))

df['tested_for_diabetes'] = np.where(df['PDIABTST'] == 1, 1,
                                    np.where(df['PDIABTST'] == 2, 0, np.nan))

# Vectorized diabetes management score
diab_cols = ['diabetes_doctor_visits', 'diabetes_eye_exam', 'uses_insulin', 'diabetes_education']
df['diabetes_management'] = df[diab_cols].sum(axis=1)
df['diabetes_management'] = df['diabetes_management'].where(
    df[diab_cols].notna().all(axis=1)
)

# Engineer Hypertension features

df['high_bp_risk'] = np.where(df['_RFHYPE5'] == 2, 1,
                             np.where(df['_RFHYPE5'] == 1, 0, np.nan))

def standardize_blood_sugar(value):
    if pd.isna(value):
        return np.nan
    if 101 <= value <= 199:
        return (value - 100) * 52  # Weekly
    elif 201 <= value <= 299:
        return (value - 200) * 12  # Monthly
    elif 301 <= value <= 399:
        return value - 300         # Yearly
    elif value == 401:
        return 0                   # Never
    return np.nan

df['blood_sugar_checks'] = df['BLDSUGAR'].apply(standardize_blood_sugar)
df['has_blood_sugar_checks'] = np.where(df['blood_sugar_checks'] >= 1, 1,
                                       np.where(df['blood_sugar_checks'] == 0, 0, np.nan))

# Vectorized hypertension management
hype_cols = ['on_bp_meds', 'has_blood_sugar_checks']
df['hypertension_management'] = df[hype_cols].sum(axis=1)
df['hypertension_management'] = df['hypertension_management'].where(
    df[hype_cols].notna().all(axis=1)
)

# Updated feature groups

chronic_conditions = {
    'Standard': ['BPHIGH4', 'BPMEDS', 'TOLDHI2', 'CHCKIDNY', 'CHKHEMO3', 
                 'has_high_bp', 'on_bp_meds', 'has_high_cholesterol', 'has_kidney_disease', 
                 'a1c_checks', 'chronic_conditions_standard'],
    'Diabetes': ['DIABETE3', 'PREDIAB1', 'DOCTDIAB', 'DIABEYE', 'INSULIN', 'DIABEDU', 'PDIABTST',
                 'has_diabetes', 'has_prediabetes', 'diabetes_doctor_visits', 'diabetes_eye_exam',
                 'uses_insulin', 'diabetes_education', 'tested_for_diabetes', 'diabetes_management'],
    'Hypertension': ['_RFHYPE5', 'BLDSUGAR', 'high_bp_risk', 'blood_sugar_checks', 
                     'has_blood_sugar_checks', 'hypertension_management']
}

# Summary Statistics Table (as requested)

categorical_cols = [
    'has_high_bp', 'on_bp_meds', 'has_high_cholesterol', 'has_kidney_disease',
    'a1c_checks', 'chronic_conditions_standard', 'has_diabetes', 'has_prediabetes',
    'diabetes_doctor_visits', 'diabetes_eye_exam', 'uses_insulin', 
    'diabetes_education', 'tested_for_diabetes', 'diabetes_management',
    'high_bp_risk', 'has_blood_sugar_checks', 'hypertension_management'
]
categorical_cols = [col for col in categorical_cols if col in df.columns]

print("\n=== Non-null Counts & Proportions ===")
for col in categorical_cols:
    non_null = df[col].notna().sum()
    if non_null == 0:
        print(f"{col:25} | Valid: {non_null:7,} | No valid data")
        continue
    yes = (df[col] == 1).sum()
    no = (df[col] == 0).sum()
    pct_yes = 100 * yes / non_null
    pct_no = 100 * no / non_null
    print(f"{col:25} | Valid: {non_null:7,} | Yes: {yes:6,} ({pct_yes:5.1f}%) | No: {no:6,} ({pct_no:5.1f}%)")

# Visualization

plt.style.use('seaborn-v0_8')

# Bar plots for binary/categorical features
n_cat = len(categorical_cols)
if n_cat > 0:
    n_cols_plot = 3
    n_rows = (n_cat + n_cols_plot - 1) // n_cols_plot
    fig, axes = plt.subplots(n_rows, n_cols_plot, figsize=(18, 5 * n_rows))
    axes = axes.flatten() if n_cat > 1 else [axes]

    for i, col in enumerate(categorical_cols):
        ax = axes[i]
        temp = df[col].dropna()
        if len(temp) == 0:
            ax.text(0.5, 0.5, 'No data', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'{col} (empty)')
        else:
            counts = temp.value_counts().reindex([0, 1], fill_value=0)
            bars = ax.bar(['No (0)', 'Yes (1)'], counts.values, color=['#1f77b4', '#d62728'])
            ax.set_title(f'{col}\n(Valid: {int(counts.sum())})')
            ax.set_ylabel('Count')
            total = counts.sum()
            for bar, count in zip(bars, counts.values):
                if count > 0:
                    pct = 100 * count / total
                    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + max(counts) * 0.01,
                            f'{int(count):,}\n({pct:.1f}%)', ha='center', va='bottom')
    
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
    
    plt.suptitle('Binary Chronic Condition Indicators', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

numeric_cols = ['blood_sugar_checks', 'DOCTDIAB', 'CHKHEMO3']
numeric_cols = [col for col in numeric_cols if col in df.columns]

if numeric_cols:
    n_num = len(numeric_cols)
    fig, axes = plt.subplots(1, n_num, figsize=(6 * n_num, 6))
    if n_num == 1:
        axes = [axes]
    for i, col in enumerate(numeric_cols):
        data = df[col].dropna()
        if len(data) == 0:
            axes[i].text(0.5, 0.5, 'No data', ha='center', va='center', transform=axes[i].transAxes)
            axes[i].set_title(f'{col} (empty)')
        else:
            sns.boxplot(y=data, ax=axes[i])
            axes[i].set_title(f'{col}\n(n={len(data):,})')
            axes[i].set_ylabel(col)
    plt.suptitle('Distribution of Numeric Chronic Condition Features', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

```

```python

# Load dataset
data_path = '../data/processed/BRFSS_11.csv'
df = pd.read_csv(data_path)

# Define Chronic Conditions & Screening features
chronic_screening_features = {
    'Arthritis': {
        'Standard': ['ARTHDIS2', 'ARTHEDU', 'ARTHEXER', 'ARTHSOCL', 'ARTHWGT', 'ARTTODAY', '_DRDXAR1'],
        'Activity_Impact': ['JOINPAIN']
    },
    'Asthma': {
        'Standard': ['ASTHMED3', 'ASTHNOW', 'ASYMPTOM', 'ASACTLIM', 'ASATTACK',
                     'ASDRVIST', 'ASERVIST', 'ASINHALR', 'ASNOSLEP', 'ASPUNSAF', 'ASRCHKUP'],
        'Other_Asthma_Features': ['_ASTHMS1', '_LTASTH1', '_CASTHM1']
    },
    'Cancer_Screening': {
        'Standard': ['PROFEXAM', 'PSATEST1', 'PSATIME', 'HADMAM', 'HADPAP2', 'HADSGCO1',
                     'HADHYST2', 'LASTPAP2', 'LASTSIG3', 'LENGEXAM'],
        'Other_Cancer': ['CHCSCNCR', 'CHCOCNCR']
    },
    'HPV_Sexual_Health': {
        'Standard': ['HIVTST6', 'HIVTSTD3', '_AIDTST3',  'HPVADVC2'],
        'Other': ['PREGNANT']
    }
}

# Flatten & keep only features present in dataset
all_chronic_features = (
    chronic_screening_features['Arthritis']['Standard']
    + chronic_screening_features['Arthritis']['Activity_Impact']
    + chronic_screening_features['Asthma']['Standard']
    + chronic_screening_features['Asthma']['Other_Asthma_Features']
    + chronic_screening_features['Cancer_Screening']['Standard']
    + chronic_screening_features['Cancer_Screening']['Other_Cancer']
    + chronic_screening_features['HPV_Sexual_Health']['Standard']
    + chronic_screening_features['HPV_Sexual_Health']['Other']
)
all_chronic_features = df.columns.intersection(all_chronic_features).tolist()

if not all_chronic_features:
    raise ValueError("No chronic conditions/screening features found in the dataset.")

# Clean BRFSS special codes

def clean_brfss_chronic(series):
    name = series.name
    if name in ['ARTHDIS2','ARTHEDU','ARTHEXER','ARTHSOCL','ARTHWGT','ARTTODAY','_DRDXAR1',
                'ASTHMED3','ASTHNOW','ASACTLIM','ASATTACK','ASDRVIST','ASERVIST','ASINHALR',
                'ASPUNSAF','ASRCHKUP','PROFEXAM','PSATEST1','HADMAM','HADPAP2','HADSGCO1',
                'HADHYST2','HIVTST6','HPVADVC2','PREGNANT']:
        return series.replace({7: np.nan, 9: np.nan})
    elif name in ['JOINPAIN','ASYMPTOM','ASNOSLEP','LENGEXAM']:
        return series.replace({77: np.nan, 99: np.nan, 88: 0})
    elif name in ['PSATIME','LASTPAP2','LASTSIG3']:
        return series.replace({77: np.nan, 99: np.nan})
    elif name in ['_ASTHMS1','_LTASTH1','_CASTHM1','_AIDTST3','CHCSCNCR','CHCOCNCR']:
        return series.replace({9: np.nan})
    return series

df[all_chronic_features] = df[all_chronic_features].apply(clean_brfss_chronic)

# Handle BRFSS zero placeholder in JOINPAIN
if 'JOINPAIN' in df.columns:
    df['JOINPAIN'] = df['JOINPAIN'].replace({5.397605e-79: 0})

# Remove all-NaN features
valid_features = [col for col in all_chronic_features if df[col].notna().any()]
if not valid_features:
    raise ValueError("All chronic conditions/screening features are NaN after cleaning.")

#  Engineer Arthritis Features 

# Arthritis diagnosis
df['arthritis_diagnosed'] = np.where(df['ARTHDIS2'] == 1, 1, 
                                    np.where(df['ARTHDIS2'] == 2, 0, np.nan))

# Arthritis impact (0-10 pain scale)
df['join_pain_level'] = df['JOINPAIN']

# Arthritis management
df['arthritis_education'] = np.where(df['ARTHEDU'] == 1, 1, 
                                    np.where(df['ARTHEDU'] == 2, 0, np.nan))
df['arthritis_exercise'] = np.where(df['ARTHEXER'] == 1, 1, 
                                   np.where(df['ARTHEXER'] == 2, 0, np.nan))
df['arthritis_social_limit'] = np.where(df['ARTHSOCL'] == 1, 1, 
                                       np.where(df['ARTHSOCL'].isin([2, 3]), 0, np.nan))

# Engineer Asthma Features 

# Current asthma status
df['current_asthma'] = np.where(df['_ASTHMS1'] == 1, 1, 
                               np.where(df['_ASTHMS1'].isin([2, 3]), 0, np.nan))

# Ever diagnosed with asthma
df['ever_asthma'] = np.where(df['_LTASTH1'] == 1, 1, 
                            np.where(df['_LTASTH1'] == 2, 0, np.nan))

# Child has asthma
df['child_has_asthma'] = np.where(df['_CASTHM1'] == 1, 1, 
                                 np.where(df['_CASTHM1'] == 2, 0, np.nan))

# Current asthma symptoms
df['asthma_now'] = np.where(df['ASTHNOW'] == 1, 1, 
                           np.where(df['ASTHNOW'] == 2, 0, np.nan))

# Asthma control indicators
df['asthma_medication'] = np.where(df['ASTHMED3'] == 1, 1, 
                                  np.where(df['ASTHMED3'] == 2, 0, np.nan))
df['asthma_attack_past_year'] = np.where(df['ASATTACK'] == 1, 1, 
                                        np.where(df['ASATTACK'] == 2, 0, np.nan))

# Engineer Cancer Screening Features 

# Mammogram screening
df['ever_mammogram'] = np.where(df['HADMAM'] == 1, 1, 
                               np.where(df['HADMAM'] == 2, 0, np.nan))
# LASTPAP2: 1=Within past year, 2=Within past 2 years, 3=Within past 5 years, 4=5+ years, 5=Never, 7=Don't know
df['recent_mammogram'] = np.where(df['LASTPAP2'].isin([1, 2, 3]), 1, 
                                 np.where(df['LASTPAP2'].isin([4, 5, 7]), 0, np.nan))

# Pap test screening
df['ever_pap_test'] = np.where(df['HADPAP2'] == 1, 1, 
                              np.where(df['HADPAP2'] == 2, 0, np.nan))
# LASTSIG3: Same coding as LASTPAP2
df['recent_pap_test'] = np.where(df['LASTSIG3'].isin([1, 2, 3]), 1, 
                                np.where(df['LASTSIG3'].isin([4, 5, 6, 7]), 0, np.nan))

# PSA test
df['ever_psa_test'] = np.where(df['PSATEST1'] == 1, 1, 
                              np.where(df['PSATEST1'] == 2, 0, np.nan))

# Cancer history
df['skin_cancer'] = np.where(df['CHCSCNCR'] == 1, 1, 
                            np.where(df['CHCSCNCR'] == 2, 0, np.nan))
df['other_cancer'] = np.where(df['CHCOCNCR'] == 1, 1, 
                             np.where(df['CHCOCNCR'] == 2, 0, np.nan))

# Engineer HPV/Sexual Health Features 

# HIV testing
df['ever_hiv_test'] = np.where(df['HIVTST6'] == 1, 1, 
                              np.where(df['HIVTST6'] == 2, 0, np.nan))

# Extract year from HIV test date (HIVTSTD3: MMYYYY or YYYY)
def extract_hiv_test_year(value):
    if pd.isna(value) or value == 999999:
        return np.nan
    year = value % 10000
    if 1900 <= year <= 2025:
        return float(year)
    return np.nan

df['hiv_test_year'] = df['HIVTSTD3'].apply(extract_hiv_test_year)


# Pregnancy status
df['is_pregnant'] = np.where(df['PREGNANT'] == 1, 1, 
                            np.where(df['PREGNANT'] == 2, 0, np.nan))

# Updated feature groups

engineered_features = {
    'Arthritis': ['arthritis_diagnosed', 'join_pain_level', 'arthritis_education', 
                  'arthritis_exercise', 'arthritis_social_limit'],
    'Asthma': ['current_asthma', 'ever_asthma', 'child_has_asthma', 'asthma_now',
               'asthma_medication', 'asthma_attack_past_year'],
    'Cancer_Screening': ['ever_mammogram', 'recent_mammogram', 'ever_pap_test', 
                         'recent_pap_test', 'ever_psa_test', 'skin_cancer', 'other_cancer'],
    'HPV_Sexual_Health': ['ever_hiv_test', 'hiv_test_year', 'hpv_vaccine', 'is_pregnant']
}

# Summary Statistics Table

categorical_cols = [
    'arthritis_diagnosed', 'arthritis_education', 'arthritis_exercise', 
    'current_asthma', 'ever_asthma', 'child_has_asthma', 'asthma_now',
    'asthma_medication', 'asthma_attack_past_year',
    'ever_mammogram', 'recent_mammogram', 'ever_pap_test', 'recent_pap_test',
    'ever_psa_test', 'skin_cancer', 'other_cancer',
    'ever_hiv_test', 'hpv_vaccine', 'is_pregnant'
]
categorical_cols = [col for col in categorical_cols if col in df.columns]

print("\n=== Non-null Counts & Proportions ===")
for col in categorical_cols:
    non_null = df[col].notna().sum()
    if non_null == 0:
        print(f"{col:25} | Valid: {non_null:7,} | No valid data")
        continue
    yes = (df[col] == 1).sum()
    no = (df[col] == 0).sum()
    pct_yes = 100 * yes / non_null
    pct_no = 100 * no / non_null
    print(f"{col:25} | Valid: {non_null:7,} | Yes: {yes:6,} ({pct_yes:5.1f}%) | No: {no:6,} ({pct_no:5.1f}%)")

# Visualization

plt.style.use('seaborn-v0_8')

# Bar plots for binary/categorical features
if categorical_cols:
    n_cols_plot = 3
    n_rows = (len(categorical_cols) + n_cols_plot - 1) // n_cols_plot
    fig, axes = plt.subplots(n_rows, n_cols_plot, figsize=(18, 5 * n_rows))
    axes = axes.flatten() if len(categorical_cols) > 1 else [axes]

    for i, col in enumerate(categorical_cols):
        ax = axes[i]
        temp = df[col].dropna()
        if len(temp) == 0:
            ax.text(0.5, 0.5, 'No data', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'{col} (empty)')
        else:
            counts = temp.value_counts().reindex([0, 1], fill_value=0)
            bars = ax.bar(['No (0)', 'Yes (1)'], counts.values, color=['#1f77b4', '#d62728'])
            ax.set_title(f'{col}\n(Valid: {int(counts.sum())})')
            ax.set_ylabel('Count')
            total = counts.sum()
            for bar, count in zip(bars, counts.values):
                if count > 0:
                    pct = 100 * count / total
                    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + max(counts) * 0.01,
                            f'{int(count):,}\n({pct:.1f}%)', ha='center', va='bottom')
    
    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
    
    plt.suptitle('Chronic Conditions & Screening Indicators', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# Box plots for numeric features
numeric_cols = ['join_pain_level', 'hiv_test_year']
numeric_cols = [col for col in numeric_cols if col in df.columns]

if numeric_cols:
    n_cols_plot = 2
    n_rows = (len(numeric_cols) + n_cols_plot - 1) // n_cols_plot
    fig, axes = plt.subplots(n_rows, n_cols_plot, figsize=(12, 5 * n_rows))
    axes = axes.flatten() if len(numeric_cols) > 1 else [axes]

    for i, col in enumerate(numeric_cols):
        ax = axes[i]
        data = df[col].dropna()
        if len(data) == 0:
            ax.text(0.5, 0.5, 'No data', ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'{col} (empty)')
        else:
            sns.boxplot(y=data, ax=ax)
            ax.set_title(f'{col}\n(n={len(data):,})')
            ax.set_ylabel(col)
    
    # Hide unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
    
    plt.suptitle('Numeric Chronic Condition Features', fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

```

```python
# Load dataset
data_path = '../data/processed/BRFSS_12.csv'
df = pd.read_csv(data_path)

# Define Income, Education, Demographics features
socio_demo_features = {
    'Income_Education': {
        'Income': ['INCOME2', '_INCOMG'],
        'Education': ['EDUCA', '_EDUCAG']
    },
    'Demographics': {
        'Basic': ['SEX', 'RCSRLTN2', 'MARITAL', 'CHILDREN',
                  'NUMWOMEN', 'NUMMEN', 'NUMADULT', 'CADULT',
                  'RCSGENDR', '_CHLDCNT']
    }
}

# Flatten & keep only features present in dataset
all_socio_features = (
    socio_demo_features['Income_Education']['Income']
    + socio_demo_features['Income_Education']['Education']
    + socio_demo_features['Demographics']['Basic']
)
all_socio_features = df.columns.intersection(all_socio_features).tolist()

if not all_socio_features:
    raise ValueError("No socioeconomic/demographic features found in dataset.")

# Cleaning function
def clean_brfss_socio(series):
    name = series.name

    # Income (INCOME2, _INCOMG): codes 77=Don't know, 99=Refused → NaN
    if name in ['INCOME2', '_INCOMG']:
        series = series.replace({77: np.nan, 99: np.nan})

    # Education (EDUCA, _EDUCAG): 9=Refused → NaN
    elif name in ['EDUCA', '_EDUCAG']:
        series = series.replace({9: np.nan})

    # Demographics categorical (SEX, MARITAL, RCSRLTN2, RCSGENDR): 9=Missing
    elif name in ['SEX', 'MARITAL', 'RCSRLTN2', 'RCSGENDR']:
        series = series.replace({9: np.nan})

    # Children counts (CHILDREN, NUMWOMEN, NUMMEN, NUMADULT, CADULT, _CHLDCNT)
    # Special missing values: 88=None, 99=Refused
    elif name in ['CHILDREN', 'NUMWOMEN', 'NUMMEN', 'NUMADULT', 'CADULT', '_CHLDCNT']:
        series = series.replace({88: 0, 99: np.nan})

    return series

# Apply cleaning
df[all_socio_features] = df[all_socio_features].apply(clean_brfss_socio)

# Remove all-NaN features
valid_features = [col for col in all_socio_features if df[col].notna().any()]
if not valid_features:
    raise ValueError("All socioeconomic/demographic features are NaN after cleaning.")

# Split numeric vs categorical
numeric_features = [f for f in valid_features if pd.api.types.is_numeric_dtype(df[f]) and df[f].nunique() > 20]
categorical_features = [f for f in valid_features if f not in numeric_features]

# Summary statistics
print("Summary statistics for Socioeconomic & Demographic features:\n")
for feature in valid_features:
    print(f"\nFeature: {feature}")
    if feature in numeric_features:
        print(df[feature].describe())
    else:
        counts = df[feature].value_counts(dropna=False)
        percentages = (counts / len(df) * 100).round(2)
        print(pd.DataFrame({'Count': counts, 'Percentage': percentages}))

# Visualization

plt.style.use('seaborn-v0_8')

# Numeric Features
if numeric_features:
    n_cols = 3
    n_rows = (len(numeric_features) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, feature in enumerate(numeric_features):
        sns.boxplot(y=df[feature], ax=axes[i])
        axes[i].set_title(f'Boxplot of {feature}')
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, feature in enumerate(numeric_features):
        sns.histplot(df[feature].dropna(), bins=20, kde=True, ax=axes[i])
        axes[i].set_title(f'Histogram of {feature}')
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()

# Categorical Features
if categorical_features:
    n_cols = 3
    n_rows = (len(categorical_features) + n_cols - 1) // n_cols
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.flatten()
    for i, feature in enumerate(categorical_features):
        order = df[feature].value_counts(dropna=False).index
        sns.countplot(x=df[feature], order=order, ax=axes[i])
        axes[i].set_title(f'Bar Plot of {feature}')
        axes[i].tick_params(axis='x', rotation=45)
    for j in range(i+1, len(axes)):
        axes[j].set_visible(False)
    plt.tight_layout()
    plt.show()
```
