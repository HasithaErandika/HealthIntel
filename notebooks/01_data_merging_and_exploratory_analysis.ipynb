```python
# Define file paths for the datasets
file_paths = {
    2015: '../data/raw/Behavioral_Risk_Factor_Surveillance_2015.csv',
    2013: '../data/raw/Behavioral_Risk_Factor_Surveillance_2013.csv',
    2011: '../data/raw/Behavioral_Risk_Factor_Surveillance_2011.csv'
}

# Load datasets and store them in a list
dfs = []
for year, path in file_paths.items():
    df = pd.read_csv(path)
    df['YEAR'] = year  # Add YEAR column
    dfs.append(df)

# Find common columns across all datasets
common_columns = list(set.intersection(*[set(df.columns) for df in dfs]))

# Subset each DataFrame to common columns and store in a list
dfs_common = [df[common_columns + ['YEAR']].copy() for df in dfs]

# Concatenate all DataFrames vertically
merged_df = pd.concat(dfs_common, ignore_index=True)

# Save the merged dataset
merged_df.to_csv('../data/raw/Behavioral_Risk_Factor_Surveillance_2011_2013_2015.csv', index=False)

# Print the shape of the merged dataset
print(f"Merged dataset shape: {merged_df.shape}")
```

```python

# Calculate total and percentage of null values for each column
null_counts = merged_df.isnull().sum()
null_percentages = (null_counts / len(merged_df)) * 100

# reate a DataFrame to summarize null values
null_summary = pd.DataFrame({
    'Null Count': null_counts,
    'Null Percentage': null_percentages
})

# Sort by percentage of null values (descending) to highlight columns with most missing data
null_summary = null_summary.sort_values(by='Null Percentage', ascending=False)

# Display columns with any null values
print("Columns with Null Values (Sorted by Percentage):")
print(null_summary[null_summary['Null Count'] > 0])

# Summary statistics of null values
print("\nSummary of Null Values:")
print(f"Total columns with null values: {len(null_summary[null_summary['Null Count'] > 0])}")
print(f"Total rows in dataset: {len(merged_df)}")
print(f"Total null values across all columns: {null_counts.sum()}")

# Visualize top 10 columns with highest null percentages

top_nulls = null_summary[null_summary['Null Count'] > 0].head(10)
plt.figure(figsize=(10, 6))
plt.bar(top_nulls.index, top_nulls['Null Percentage'])
plt.xticks(rotation=45, ha='right')
plt.xlabel('Column Name')
plt.ylabel('Percentage of Null Values')
plt.title('Top 10 Columns with Highest Null Values')
plt.tight_layout()
plt.show()

print("Columns:", merged_df.columns.tolist())
```

```python
# Drop columns with >65% missing
threshold = 65  
missing_percent = (merged_df.isna().sum() / len(merged_df)) * 100

# Columns to drop
high_missing_cols = missing_percent[missing_percent > threshold].index.tolist()

print(f"Columns with >{threshold}% missing: {len(high_missing_cols)}")
print(high_missing_cols)

# Drop them
df_cleaned = merged_df.drop(columns=high_missing_cols)

print(f"Original shape: {merged_df.shape}")
print(f"New shape after dropping: {df_cleaned.shape}")
```
